```python
# ============================================================
# CELL 3: CONFIGURATION + DATASET DISCOVERY & ANALYSIS (FIXED v3)
# ============================================================

# ============================================================
# 3.1 CONFIGURATION
# ============================================================

from dataclasses import dataclass, field
from typing import List, Tuple, Optional

@dataclass
class CFG:
    """TopoEyeNet Experiment Configuration"""

    EXPERIMENT_NAME: str = "topoeyenet_v1"
    SEED: int = 42

    RETINOMIX_IMAGE_DIR: str = "/kaggle/input/datasets/parikshakaur/retinomix-5/RETINOMIX-5/images"
    RETINOMIX_MASK_DIR: str = "/kaggle/input/datasets/parikshakaur/retinomix-5/RETINOMIX-5/mask"

    DRIVE_TRAIN_IMAGE_DIR: str = "/kaggle/input/datasets/andrewmvd/drive-digital-retinal-images-for-vessel-extraction/DRIVE/training/images"
    DRIVE_TRAIN_MASK_DIR: str = "/kaggle/input/datasets/andrewmvd/drive-digital-retinal-images-for-vessel-extraction/DRIVE/training/1st_manual"
    DRIVE_TRAIN_FOV_DIR: str = "/kaggle/input/datasets/andrewmvd/drive-digital-retinal-images-for-vessel-extraction/DRIVE/training/mask"

    DRIVE_TEST_IMAGE_DIR: str = "/kaggle/input/datasets/andrewmvd/drive-digital-retinal-images-for-vessel-extraction/DRIVE/test/images"
    DRIVE_TEST_MASK_DIR: str = "/kaggle/input/datasets/andrewmvd/drive-digital-retinal-images-for-vessel-extraction/DRIVE/test/mask"  # FIXED: test masks are here
    DRIVE_TEST_FOV_DIR: str = "/kaggle/input/datasets/andrewmvd/drive-digital-retinal-images-for-vessel-extraction/DRIVE/test/mask"

    EYEPACS_IMAGE_DIR: str = "/kaggle/input/datasets/dreamer07/eyepacs/data/data"
    EYEPACS_CSV: str = "/kaggle/input/datasets/dreamer07/eyepacs/trainLabels.csv/trainLabels.csv"

    OUTPUT_DIR: str = "/kaggle/working/output"
    CHECKPOINT_DIR: str = "/kaggle/working/checkpoints"
    LOG_DIR: str = "/kaggle/working/logs"
    PRECOMPUTED_DIR: str = "/kaggle/working/precomputed"
    FIGURES_DIR: str = "/kaggle/working/figures"

    IMG_SIZE: int = 512
    IN_CHANNELS: int = 3
    NUM_CLASSES: int = 1

    ENCODER: str = "efficientnet-b3"
    ENCODER_WEIGHTS: str = "imagenet"
    DECODER: str = "unet"

    USE_MASK_HEAD: bool = True
    USE_SKELETON_HEAD: bool = True
    USE_WIDTH_HEAD: bool = False
    USE_ORIENTATION_HEAD: bool = False
    USE_JUNCTION_HEAD: bool = False

    BATCH_SIZE: int = 4
    NUM_WORKERS: int = 2
    EPOCHS: int = 100
    LEARNING_RATE: float = 1e-4
    WEIGHT_DECAY: float = 1e-5

    LAMBDA_MASK: float = 1.0
    LAMBDA_SKELETON: float = 0.5
    LAMBDA_CONSISTENCY: float = 0.3

    OPTIMIZER: str = "adamw"
    SCHEDULER: str = "cosine"
    T_MAX: int = 100
    ETA_MIN: float = 1e-6

    AUG_PROB: float = 0.5
    USE_CUTMIX: bool = False

    USE_AMP: bool = True
    USE_DATAPARALLEL: bool = True

    VAL_FREQ: int = 1
    SAVE_BEST_ONLY: bool = True
    EARLY_STOPPING_PATIENCE: int = 15

    PRECOMPUTE_REPRESENTATIONS: bool = True
    SKELETON_METHOD: str = "zhang"
    MIN_VESSEL_SIZE: int = 20

    USE_WANDB: bool = False
    WANDB_PROJECT: str = "topoeyenet"
    LOG_INTERVAL: int = 10

    GRADIENT_ACCUMULATION_STEPS: int = 1
    CLEAR_CACHE_EVERY_N_BATCHES: int = 50

    def create_dirs(self):
        for d in [self.OUTPUT_DIR, self.CHECKPOINT_DIR, self.LOG_DIR,
                  self.PRECOMPUTED_DIR, self.FIGURES_DIR]:
            os.makedirs(d, exist_ok=True)

    def to_dict(self):
        return {k: v for k, v in self.__dict__.items() if not k.startswith('_')}

cfg = CFG()
cfg.create_dirs()
seed_everything(cfg.SEED)

print("=" * 60)
print("CONFIGURATION")
print("=" * 60)
for k, v in cfg.to_dict().items():
    print(f"  {k}: {v}")
print("=" * 60)
print(f"Effective batch size: {cfg.BATCH_SIZE * max(1, device_count) * cfg.GRADIENT_ACCUMULATION_STEPS}")

# ============================================================
# 3.2 DATASET DISCOVERY & ANALYSIS
# ============================================================

def discover_images(directory: str, extensions=('.png', '.jpg', '.jpeg', '.tif', '.tiff', '.bmp', '.gif')) -> List[str]:
    if not os.path.exists(directory):
        return []
    files = []
    for root, _, filenames in os.walk(directory):
        for f in filenames:
            if f.lower().endswith(extensions):
                files.append(os.path.join(root, f))
    return sorted(files)

def get_image_info(path: str) -> dict:
    try:
        with Image.open(path) as img:
            return {'path': path, 'width': img.size[0], 'height': img.size[1],
                    'mode': img.mode, 'ok': True}
    except Exception as e:
        return {'path': path, 'width': 0, 'height': 0, 'mode': 'ERROR', 'ok': False, 'error': str(e)}

def analyze_mask(mask_path: str) -> dict:
    try:
        mask = np.array(Image.open(mask_path).convert('L'))
        if mask.max() > 1:
            mask = (mask > 127).astype(np.uint8)
        vessel_pixels = mask.sum()
        total = mask.size
        vessel_pct = 100.0 * vessel_pixels / total
        labeled = measure.label(mask, connectivity=2)
        return {
            'vessel_pixels': int(vessel_pixels), 'total_pixels': int(total),
            'vessel_pct': round(vessel_pct, 3), 'num_components': int(labeled.max()), 'ok': True
        }
    except Exception as e:
        return {'vessel_pixels': 0, 'total_pixels': 0, 'vessel_pct': 0,
                'num_components': 0, 'ok': False, 'error': str(e)}

def extract_id(path: str) -> str:
    """
    Source- AND split-aware ID extraction (FIXED v3).

    Handles two distinct filename conventions found across the merged
    datasets:

      - RetinoMix-style: shared trailing digit + literal "image"/"mask"
        word, e.g. 'CHASE_train_image_01.png' / 'CHASE_train_mask_01.png'.
        Stripping only the words "image"/"mask" keeps source + split +
        id intact as the key, so different sources and different splits
        never collide:
          'CHASE_train_image_01.png' -> 'chase_train_01'
          'CHASE_train_mask_01.png'  -> 'chase_train_01'   (matches)
          'CHASE_test_image_01.png'  -> 'chase_test_01'    (does NOT
                                          match the train version)

      - DRIVE-original-style: shared LEADING digit, no shared word at
        all, e.g. '21_training.tif' (image) / '21_manual1.gif' (mask).
        This convention is checked FIRST since it can't be recovered by
        word-stripping (there's no "image"/"mask" substring in either
        filename):
          '21_training.tif' -> 'leading_21'
          '21_manual1.gif'  -> 'leading_21'                (matches)
    """
    basename = os.path.splitext(os.path.basename(path))[0].lower()

    leading_match = re.match(r'^(\d+)', basename)
    if leading_match:
        return f"leading_{leading_match.group(1)}"

    key = re.sub(r'(image|mask)', '', basename)
    key = re.sub(r'_+', '_', key).strip('_')
    return key


def match_images_to_masks(image_paths, mask_paths):
    """
    Match images to masks by extracting a source- and split-aware ID
    from filenames. Returns (pairs, unmatched) -- same signature as
    the original.
    """
    mask_lookup = {}
    for mp in mask_paths:
        mask_lookup[extract_id(mp)] = mp

    pairs, unmatched = [], []
    for ip in image_paths:
        iid = extract_id(ip)
        if iid in mask_lookup:
            pairs.append((ip, mask_lookup[iid]))
        else:
            unmatched.append(ip)

    return pairs, unmatched

class DatasetAnalyzer:
    def __init__(self, name: str, image_dir: str, mask_dir: str = None,
                 fov_dir: str = None, csv_path: str = None):
        self.name = name
        self.image_dir = image_dir
        self.mask_dir = mask_dir
        self.fov_dir = fov_dir
        self.csv_path = csv_path
        self.pairs = []
        self.stats = {}

    def analyze(self):
        print(f"\n{'='*60}")
        print(f"DATASET: {self.name}")
        print(f"{'='*60}")

        image_paths = discover_images(self.image_dir)
        print(f"  Images found: {len(image_paths)}")
        if not image_paths:
            print(f"  WARNING: No images in {self.image_dir}")
            return self.stats

        sample = get_image_info(image_paths[0])
        print(f"  Sample format: {sample['mode']} | {sample['width']}x{sample['height']}")

        mask_paths = []
        if self.mask_dir and os.path.exists(self.mask_dir):
            mask_paths = discover_images(self.mask_dir)
            print(f"  Masks found: {len(mask_paths)}")
        else:
            print(f"  Masks: NOT AVAILABLE")

        if mask_paths:
            pairs, unmatched = match_images_to_masks(image_paths, mask_paths)
            self.pairs = pairs
            print(f"  Matched pairs: {len(pairs)}")
            if unmatched:
                print(f"  Unmatched images: {len(unmatched)}")
                for u in unmatched[:3]:
                    print(f"    - {os.path.basename(u)}")
        else:
            self.pairs = [(ip, None) for ip in image_paths]

        resolutions = []
        for ip, _ in self.pairs[:50]:
            info = get_image_info(ip)
            if info['ok']:
                resolutions.append((info['width'], info['height']))
        if resolutions:
            w, h = zip(*resolutions)
            print(f"  Resolution range: {min(w)}x{min(h)} to {max(w)}x{max(h)}")
            unique = sorted(set(resolutions))
            print(f"  Unique resolutions ({len(unique)}): {unique[:5]}")

        if mask_paths and self.pairs:
            v_pcts, comps = [], []
            for _, mp in self.pairs[:20]:
                if mp:
                    s = analyze_mask(mp)
                    if s['ok']:
                        v_pcts.append(s['vessel_pct'])
                        comps.append(s['num_components'])
            if v_pcts:
                print(f"  Vessel % (sample): {min(v_pcts):.2f}% - {max(v_pcts):.2f}% | mean={np.mean(v_pcts):.2f}%")
                print(f"  Components (sample): mean={np.mean(comps):.1f}")

        if self.csv_path and os.path.exists(self.csv_path):
            df = pd.read_csv(self.csv_path)
            print(f"  CSV rows: {len(df)}")
            if 'level' in df.columns:
                print(f"  Grade distribution: {df['level'].value_counts().sort_index().to_dict()}")

        self.stats = {
            'name': self.name, 'num_images': len(image_paths),
            'num_masks': len(mask_paths), 'num_pairs': len(self.pairs),
            'has_masks': len(mask_paths) > 0,
            'sample_resolution': resolutions[0] if resolutions else None,
        }
        return self.stats

# ============================================================
# 3.3 RUN ANALYSIS
# ============================================================

print("\nStarting dataset analysis...")
datasets = []

retinomix = DatasetAnalyzer("RetinoMix-5", cfg.RETINOMIX_IMAGE_DIR, cfg.RETINOMIX_MASK_DIR)
retinomix.analyze(); datasets.append(retinomix)

drive_train = DatasetAnalyzer("DRIVE (Training)", cfg.DRIVE_TRAIN_IMAGE_DIR,
                              cfg.DRIVE_TRAIN_MASK_DIR, cfg.DRIVE_TRAIN_FOV_DIR)
drive_train.analyze(); datasets.append(drive_train)

drive_test = DatasetAnalyzer("DRIVE (Test)", cfg.DRIVE_TEST_IMAGE_DIR,
                             cfg.DRIVE_TEST_MASK_DIR, cfg.DRIVE_TEST_FOV_DIR)
drive_test.analyze(); datasets.append(drive_test)

eyepacs = DatasetAnalyzer("EyePACS", cfg.EYEPACS_IMAGE_DIR, csv_path=cfg.EYEPACS_CSV)
eyepacs.analyze(); datasets.append(eyepacs)

# Summary
print(f"\n{'='*60}")
print("DATASET SUMMARY")
print(f"{'='*60}")
summary = []
for ds in datasets:
    s = ds.stats
    summary.append({
        'Dataset': s['name'], 'Images': s['num_images'], 'Masks': s['num_masks'],
        'Pairs': s['num_pairs'], 'Supervised': 'Yes' if s['has_masks'] else 'No',
        'Resolution': str(s['sample_resolution']) if s['sample_resolution'] else 'N/A'
    })
summary_df = pd.DataFrame(summary)
print(summary_df.to_string(index=False))
summary_df.to_csv(os.path.join(cfg.LOG_DIR, 'dataset_summary.csv'), index=False)
print(f"\nSaved: {cfg.LOG_DIR}/dataset_summary.csv")

# ============================================================
# 3.4 VISUALIZE SAMPLE PAIRS
# ============================================================

def visualize_samples(dataset: DatasetAnalyzer, n: int = 3):
    pairs = dataset.pairs[:n]
    if not pairs:
        return
    cols = 3 if pairs[0][1] else 2
    fig, axes = plt.subplots(n, cols, figsize=(12, 4*n))
    if n == 1:
        axes = axes.reshape(1, -1)

    for idx, (img_path, mask_path) in enumerate(pairs):
        img = np.array(Image.open(img_path).convert('RGB'))
        h, w = img.shape[:2]

        if mask_path and os.path.exists(mask_path):
            mask = np.array(Image.open(mask_path).convert('L'))
            # Resize mask to match image if needed
            if mask.shape[0] != h or mask.shape[1] != w:
                mask = cv2.resize(mask, (w, h), interpolation=cv2.INTER_NEAREST)
            mask_bin = (mask > 127).astype(np.uint8)
        else:
            mask = None
            mask_bin = None

        # Now resize image for display if too large
        if max(h, w) > 1024:
            scale = 1024 / max(h, w)
            new_w, new_h = int(w*scale), int(h*scale)
            img = cv2.resize(img, (new_w, new_h))
            if mask_bin is not None:
                mask_bin = cv2.resize(mask_bin, (new_w, new_h), interpolation=cv2.INTER_NEAREST)

        axes[idx, 0].imshow(img)
        axes[idx, 0].set_title(f"Input\n{os.path.basename(img_path)}", fontsize=8)
        axes[idx, 0].axis('off')

        if mask_bin is not None:
            axes[idx, 1].imshow(mask_bin, cmap='gray')
            axes[idx, 1].set_title(f"Mask\n{os.path.basename(mask_path)}", fontsize=8)
            axes[idx, 1].axis('off')

            overlay = img.copy()
            overlay[mask_bin > 0] = [255, 0, 0]
            axes[idx, 2].imshow(overlay)
            axes[idx, 2].set_title("Overlay", fontsize=8)
            axes[idx, 2].axis('off')
        else:
            axes[idx, 1].text(0.5, 0.5, "No mask", ha='center', va='center', fontsize=12)
            axes[idx, 1].axis('off')
            if cols > 2:
                axes[idx, 2].axis('off')

    plt.suptitle(f"Dataset: {dataset.name}", fontsize=12, fontweight='bold')
    plt.tight_layout()
    save_path = os.path.join(cfg.FIGURES_DIR, f"samples_{dataset.name.replace(' ', '_').replace('(', '').replace(')', '')}.png")
    plt.savefig(save_path, dpi=150, bbox_inches='tight')
    print(f"Saved figure: {save_path}")
    plt.close()

for ds in datasets:
    if ds.stats['has_masks'] and ds.pairs:
        visualize_samples(ds, n=min(3, len(ds.pairs)))

# Memory check
print(f"\n{'='*60}")
print("MEMORY STATUS")
print(f"{'='*60}")
get_gpu_memory()

output_size = sum(os.path.getsize(os.path.join(dp, f))
                  for dp, dn, filenames in os.walk(cfg.OUTPUT_DIR)
                  for f in filenames if os.path.exists(os.path.join(dp, f))) / 1e9
print(f"  Output dir: {output_size:.3f} GB / 20 GB limit")

print(f"\n{'='*60}")
print("CELL 3 COMPLETE")
print(f"{'='*60}")
```

```python
# ============================================================
# CELL 4: REPRESENTATION GENERATOR
# ============================================================
# Precomputes multi-task supervision targets from vessel masks:
#   - Skeleton
#   - Distance Transform
#   - Orientation Map
#   - Junction Map
#   - Endpoint Map
#   - Width Map
# All saved to /kaggle/working/precomputed/ for fast training loading.
# Also generates comparison figures (shown inline + saved).

import os
import numpy as np
from PIL import Image
import cv2
from skimage.morphology import skeletonize, medial_axis, remove_small_objects
from scipy.ndimage import distance_transform_edt, gaussian_filter1d
from scipy import ndimage
import matplotlib.pyplot as plt
from pathlib import Path
import time

# ============================================================
# 4.1 CORE GENERATION FUNCTIONS
# ============================================================

def generate_skeleton(mask: np.ndarray, method: str = "zhang") -> np.ndarray:
    """
    Generate skeleton from binary vessel mask.
    Methods: 'zhang', 'lee', 'medial_axis'
    """
    mask_bin = (mask > 0).astype(np.uint8)
    
    if method == "zhang":
        skel = skeletonize(mask_bin, method='zhang')
    elif method == "lee":
        skel = skeletonize(mask_bin, method='lee')
    elif method == "medial_axis":
        skel, _ = medial_axis(mask_bin, return_distance=True)
    else:
        skel = skeletonize(mask_bin, method='zhang')
    
    return skel.astype(np.uint8)

def generate_distance_transform(mask: np.ndarray) -> np.ndarray:
    """
    Euclidean distance transform: each pixel = distance to nearest background.
    For vessels, this gives half-width at each point.
    """
    mask_bin = (mask > 0).astype(np.uint8)
    # Distance from vessel pixels to background
    dist = distance_transform_edt(mask_bin)
    return dist.astype(np.float32)

def generate_width_map(mask: np.ndarray, skeleton: np.ndarray) -> np.ndarray:
    """
    Vessel width at each pixel: 2 * distance transform on skeleton pixels.
    """
    dist = generate_distance_transform(mask)
    width = np.zeros_like(mask, dtype=np.float32)
    # Width = 2 * distance transform value on skeleton
    width[skeleton > 0] = 2 * dist[skeleton > 0]
    # Interpolate width to full vessel mask for dense supervision
    width = cv2.inpaint(
        (width * 255).astype(np.uint8),
        (mask == 0).astype(np.uint8),
        inpaintRadius=3,
        flags=cv2.INPAINT_NS
    )
    return width.astype(np.float32) / 255.0

def generate_orientation_map(mask: np.ndarray, skeleton: np.ndarray) -> np.ndarray:
    """
    Vessel orientation (angle in radians) at each skeleton pixel.
    Computed from local gradient of the distance transform.
    """
    dist = generate_distance_transform(mask)
    
    # Smooth distance transform
    dist_smooth = gaussian_filter1d(gaussian_filter1d(dist, sigma=1.0, axis=0), sigma=1.0, axis=1)
    
    # Compute gradient
    gy, gx = np.gradient(dist_smooth)
    
    # Orientation = angle perpendicular to gradient (along vessel)
    orientation = np.arctan2(-gx, gy)  # perpendicular to gradient
    
    # Only on skeleton
    orient_map = np.zeros_like(mask, dtype=np.float32)
    orient_map[skeleton > 0] = orientation[skeleton > 0]
    
    # Interpolate to full vessel
    if mask.sum() > 0:
        orient_norm = (orient_map + np.pi) / (2 * np.pi)  # normalize to [0,1] for inpainting
        orient_inpaint = cv2.inpaint(
            (orient_norm * 255).astype(np.uint8),
            ((mask == 0) | (skeleton == 0)).astype(np.uint8),
            inpaintRadius=5,
            flags=cv2.INPAINT_NS
        )
        orient_map = (orient_inpaint.astype(np.float32) / 255.0) * 2 * np.pi - np.pi
    
    return orient_map.astype(np.float32)

def generate_junction_map(skeleton: np.ndarray) -> np.ndarray:
    """
    Detect junctions (branching points) in skeleton.
    A junction pixel has > 2 neighbors in the skeleton.
    """
    from scipy.signal import convolve2d
    
    kernel = np.array([[1, 1, 1],
                       [1, 10, 1],
                       [1, 1, 1]], dtype=np.uint8)
    
    conv = convolve2d(skeleton, kernel, mode='same', boundary='fill')
    # Junction: center is skeleton (10) + 3+ neighbors
    junction = ((conv >= 13) & (skeleton > 0)).astype(np.uint8)
    
    # Dilate slightly for visibility
    junction = cv2.dilate(junction, np.ones((3, 3), np.uint8), iterations=1)
    
    return junction

def generate_endpoint_map(skeleton: np.ndarray) -> np.ndarray:
    """
    Detect endpoints (terminations) in skeleton.
    An endpoint has exactly 1 neighbor.
    """
    from scipy.signal import convolve2d
    
    kernel = np.array([[1, 1, 1],
                       [1, 10, 1],
                       [1, 1, 1]], dtype=np.uint8)
    
    conv = convolve2d(skeleton, kernel, mode='same', boundary='fill')
    # Endpoint: center is skeleton (10) + exactly 1 neighbor = 11
    endpoint = ((conv == 11) & (skeleton > 0)).astype(np.uint8)
    
    # Dilate for visibility
    endpoint = cv2.dilate(endpoint, np.ones((3, 3), np.uint8), iterations=1)
    
    return endpoint

def generate_all_representations(mask: np.ndarray, method: str = "zhang", min_size: int = 20) -> dict:
    """
    Generate all topological representations from a vessel mask.
    Returns dict with all maps.
    """
    # Clean mask: remove small components
    mask_clean = remove_small_objects(mask > 0, min_size=min_size)
    mask_clean = mask_clean.astype(np.uint8)
    
    # Skeleton
    skeleton = generate_skeleton(mask_clean, method=method)
    
    # Distance transform
    distance = generate_distance_transform(mask_clean)
    
    # Width map
    width = generate_width_map(mask_clean, skeleton)
    
    # Orientation
    orientation = generate_orientation_map(mask_clean, skeleton)
    
    # Junctions
    junction = generate_junction_map(skeleton)
    
    # Endpoints
    endpoint = generate_endpoint_map(skeleton)
    
    return {
        'mask': mask_clean,
        'skeleton': skeleton,
        'distance': distance,
        'width': width,
        'orientation': orientation,
        'junction': junction,
        'endpoint': endpoint,
    }

# ============================================================
# 4.2 VISUALIZATION (inline + saved)
# ============================================================

def visualize_representations(reps: dict, title: str = "Representations", 
                                 save_path: str = None, show_inline: bool = True):
    """
    Visualize all representations in a grid.
    Shows inline in notebook AND saves to file.
    """
    fig, axes = plt.subplots(2, 4, figsize=(16, 8))
    axes = axes.flatten()
    
    # Color maps for each representation
    displays = [
        ('mask', reps['mask'], 'gray', 'Vessel Mask'),
        ('skeleton', reps['skeleton'], 'gray', 'Skeleton'),
        ('distance', reps['distance'], 'viridis', 'Distance Transform'),
        ('width', reps['width'], 'plasma', 'Width Map'),
        ('orientation', (reps['orientation'] + np.pi) / (2 * np.pi), 'hsv', 'Orientation Map'),
        ('junction', reps['junction'], 'hot', 'Junction Map'),
        ('endpoint', reps['endpoint'], 'hot', 'Endpoint Map'),
    ]
    
    for ax, (key, data, cmap, label) in zip(axes, displays):
        im = ax.imshow(data, cmap=cmap)
        ax.set_title(label, fontsize=10)
        ax.axis('off')
        plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    
    # Hide last subplot
    axes[-1].axis('off')
    
    plt.suptitle(title, fontsize=14, fontweight='bold')
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"Saved: {save_path}")
    
    if show_inline:
        plt.show()
    else:
        plt.close()

# ============================================================
# 4.3 BATCH PROCESSING
# ============================================================

def precompute_dataset_representations(dataset: DatasetAnalyzer, 
                                        output_dir: str,
                                        method: str = "zhang",
                                        min_size: int = 20,
                                        max_samples: int = None):
    """
    Precompute representations for all masks in a dataset.
    Saves as .npz files and generates sample visualizations.
    """
    os.makedirs(output_dir, exist_ok=True)
    
    pairs = dataset.pairs[:max_samples] if max_samples else dataset.pairs
    print(f"\nPrecomputing representations for {dataset.name}: {len(pairs)} images")
    
    stats = {
        'num_processed': 0,
        'num_junctions': [],
        'num_endpoints': [],
        'mean_width': [],
        'mean_vessel_pct': [],
    }
    
    for idx, (img_path, mask_path) in enumerate(pairs):
        if mask_path is None:
            continue
        
        # Load mask
        mask = np.array(Image.open(mask_path).convert('L'))
        if mask.max() > 1:
            mask = (mask > 127).astype(np.uint8)
        
        # Generate representations
        reps = generate_all_representations(mask, method=method, min_size=min_size)
        
        # Save as compressed NPZ
        base_name = os.path.splitext(os.path.basename(mask_path))[0]
        save_path = os.path.join(output_dir, f"{base_name}_reps.npz")
        np.savez_compressed(save_path, **reps)
        
        # Stats
        stats['num_processed'] += 1
        stats['num_junctions'].append(reps['junction'].sum())
        stats['num_endpoints'].append(reps['endpoint'].sum())
        stats['mean_width'].append(reps['width'][reps['mask'] > 0].mean())
        stats['mean_vessel_pct'].append(reps['mask'].sum() / reps['mask'].size * 100)
        
        # Visualize first 3 samples
        if idx < 3:
            fig_path = os.path.join(cfg.FIGURES_DIR, f"reps_{dataset.name.replace(' ', '_')}_{idx}.png")
            visualize_representations(
                reps, 
                title=f"{dataset.name} — {os.path.basename(mask_path)}",
                save_path=fig_path,
                show_inline=True  # Show in notebook output
            )
        
        if (idx + 1) % 50 == 0:
            print(f"  Processed {idx + 1}/{len(pairs)}...")
    
    # Summary stats
    print(f"\n{'='*60}")
    print(f"PRECOMPUTATION SUMMARY: {dataset.name}")
    print(f"{'='*60}")
    print(f"  Processed: {stats['num_processed']}")
    print(f"  Avg junctions: {np.mean(stats['num_junctions']):.1f}")
    print(f"  Avg endpoints: {np.mean(stats['num_endpoints']):.1f}")
    print(f"  Avg width (px): {np.mean(stats['mean_width']):.2f}")
    print(f"  Avg vessel %: {np.mean(stats['mean_vessel_pct']):.2f}%")
    print(f"  Saved to: {output_dir}")
    
    return stats

# ============================================================
# 4.4 RUN ON ALL SUPERVISED DATASETS
# ============================================================

print("=" * 60)
print("REPRESENTATION GENERATOR")
print("=" * 60)

# Process RetinoMix-5
retinomix_reps_dir = os.path.join(cfg.PRECOMPUTED_DIR, "retinomix5")
retinomix_stats = precompute_dataset_representations(
    retinomix, 
    output_dir=retinomix_reps_dir,
    method=cfg.SKELETON_METHOD,
    min_size=cfg.MIN_VESSEL_SIZE
)

# Process DRIVE Training
drive_reps_dir = os.path.join(cfg.PRECOMPUTED_DIR, "drive_train")
drive_stats = precompute_dataset_representations(
    drive_train,
    output_dir=drive_reps_dir,
    method=cfg.SKELETON_METHOD,
    min_size=cfg.MIN_VESSEL_SIZE
)

# Check output size
precomputed_size = sum(
    os.path.getsize(os.path.join(dp, f))
    for dp, dn, filenames in os.walk(cfg.PRECOMPUTED_DIR)
    for f in filenames if os.path.exists(os.path.join(dp, f))
) / 1e9
print(f"\nPrecomputed dir size: {precomputed_size:.3f} GB")

# Memory check
print(f"\n{'='*60}")
print("MEMORY STATUS")
print(f"{'='*60}")
get_gpu_memory()

print(f"\n{'='*60}")
print("CELL 4 COMPLETE")
print(f"{'='*60}")
```

```python
# ============================================================
# CELL 5: MULTI-TASK DATASET & DATALOADER (FIXED)
# ============================================================
# Loads images + precomputed representations (mask, skeleton, width, orientation, junction)
# Applies synchronized augmentations to all targets.
# Returns: image, {mask, skeleton, width, orientation, junction, endpoint}

import os
import numpy as np
from PIL import Image
import torch
from torch.utils.data import Dataset, DataLoader
import cv2

# ============================================================
# 5.1 AUGMENTATION PIPELINE (local import check, no external flags)
# ============================================================

def get_training_augmentation(img_size: int = 512, aug_prob: float = 0.5):
    """
    Training augmentation pipeline.
    Auto-detects albumentations availability locally.
    """
    try:
        import albumentations as A
        from albumentations.pytorch import ToTensorV2
        HAS_ALBU = True
    except ImportError:
        HAS_ALBU = False
    
    if HAS_ALBU:
        transform = A.Compose([
            A.Resize(img_size, img_size),
            A.HorizontalFlip(p=aug_prob),
            A.VerticalFlip(p=aug_prob),
            A.RandomRotate90(p=aug_prob),
            A.ShiftScaleRotate(shift_limit=0.05, scale_limit=0.1, rotate_limit=15, p=aug_prob),
            A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=aug_prob),
            A.GaussNoise(var_limit=(10, 50), p=aug_prob * 0.5),
            A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
            ToTensorV2(),
        ], additional_targets={
            'mask': 'mask',
            'skeleton': 'mask',
            'width': 'mask',
            'orientation': 'mask',
            'junction': 'mask',
            'endpoint': 'mask',
        })
        return transform, 'albumentations'
    else:
        from torchvision import transforms
        transform = transforms.Compose([
            transforms.Resize((img_size, img_size)),
            transforms.RandomHorizontalFlip(p=aug_prob),
            transforms.RandomVerticalFlip(p=aug_prob),
            transforms.RandomRotation(15),
            transforms.ColorJitter(brightness=0.2, contrast=0.2),
            transforms.ToTensor(),
            transforms.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ])
        return transform, 'torchvision'

def get_validation_augmentation(img_size: int = 512):
    """Validation/test: only resize + normalize, but must declare the
    same additional_targets as training so multi-task targets
    (skeleton, width, orientation, junction, endpoint) aren't rejected."""
    try:
        import albumentations as A
        from albumentations.pytorch import ToTensorV2
        HAS_ALBU = True
    except ImportError:
        HAS_ALBU = False

    if HAS_ALBU:
        transform = A.Compose([
            A.Resize(img_size, img_size),
            A.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
            ToTensorV2(),
        ], additional_targets={
            'mask': 'mask',
            'skeleton': 'mask',
            'width': 'mask',
            'orientation': 'mask',
            'junction': 'mask',
            'endpoint': 'mask',
        })
        return transform, 'albumentations'
    else:
        from torchvision import transforms
        transform = transforms.Compose([
            transforms.Resize((img_size, img_size)),
            transforms.ToTensor(),
            transforms.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225)),
        ])
        return transform, 'torchvision'
# ============================================================
# 5.2 MULTI-TASK DATASET
# ============================================================

class RetinalMultiTaskDataset(Dataset):
    """
    Dataset that loads:
      - Fundus image
      - Precomputed representations (from .npz files)
    
    Returns:
      image: (3, H, W) tensor
      targets: dict of tensors {mask, skeleton, width, orientation, junction, endpoint}
    """
    
    def __init__(self, 
                 image_paths: list,
                 mask_paths: list,
                 precomputed_dir: str,
                 transform=None,
                 transform_type: str = 'albumentations',
                 img_size: int = 512,
                 use_cache: bool = True):
        
        self.image_paths = image_paths
        self.mask_paths = mask_paths
        self.precomputed_dir = precomputed_dir
        self.transform = transform
        self.transform_type = transform_type
        self.img_size = img_size
        self.use_cache = use_cache
        self.cache = {} if use_cache else None
        
        # Validate: check all precomputed files exist
        self.valid_indices = []
        for i, mp in enumerate(mask_paths):
            if mp is None:
                continue
            base = os.path.splitext(os.path.basename(mp))[0]
            npz_path = os.path.join(precomputed_dir, f"{base}_reps.npz")
            if os.path.exists(npz_path):
                self.valid_indices.append(i)
        
        print(f"Dataset: {len(self.valid_indices)}/{len(image_paths)} valid samples")
        print(f"  Precomputed dir: {precomputed_dir}")
        print(f"  Transform: {transform_type}")
        
    def __len__(self):
        return len(self.valid_indices)
    
    def __getitem__(self, idx):
        real_idx = self.valid_indices[idx]
        
        img_path = self.image_paths[real_idx]
        mask_path = self.mask_paths[real_idx]
        
        # Load image
        image = np.array(Image.open(img_path).convert('RGB'))
        h, w = image.shape[:2]
        
        # Load precomputed representations
        base = os.path.splitext(os.path.basename(mask_path))[0]
        npz_path = os.path.join(self.precomputed_dir, f"{base}_reps.npz")
        
        if self.use_cache and npz_path in self.cache:
            reps = self.cache[npz_path]
        else:
            reps = dict(np.load(npz_path))
            if self.use_cache:
                self.cache[npz_path] = reps
        
        # Extract targets and ensure they match image shape
        mask = reps['mask'].astype(np.float32)
        skeleton = reps['skeleton'].astype(np.float32)
        width = reps['width'].astype(np.float32)
        orientation = reps['orientation'].astype(np.float32)
        junction = reps['junction'].astype(np.float32)
        endpoint = reps['endpoint'].astype(np.float32)
        
        # CRITICAL: Resize all targets to match image shape before albumentations
        def resize_to_match(arr, target_h, target_w, interp=cv2.INTER_NEAREST):
            if arr.shape[0] != target_h or arr.shape[1] != target_w:
                return cv2.resize(arr, (target_w, target_h), interpolation=interp)
            return arr
        
        mask = resize_to_match(mask, h, w, cv2.INTER_NEAREST)
        skeleton = resize_to_match(skeleton, h, w, cv2.INTER_NEAREST)
        width = resize_to_match(width, h, w, cv2.INTER_LINEAR)
        orientation = resize_to_match(orientation, h, w, cv2.INTER_LINEAR)
        junction = resize_to_match(junction, h, w, cv2.INTER_NEAREST)
        endpoint = resize_to_match(endpoint, h, w, cv2.INTER_NEAREST)
        
        # Apply transforms
        if self.transform_type == 'albumentations' and self.transform:
            transformed = self.transform(
                image=image,
                mask=mask,
                skeleton=skeleton,
                width=width,
                orientation=orientation,
                junction=junction,
                endpoint=endpoint,
            )
            
            # ToTensorV2 converts image and all additional targets to tensors
            image_tensor = transformed['image']  # Already tensor (3, H, W)
            
            def to_tensor(v):
                """Convert to tensor, handling both numpy and tensor inputs."""
                if isinstance(v, torch.Tensor):
                    return v.unsqueeze(0).float()
                else:
                    return torch.from_numpy(v).unsqueeze(0).float()
            
            targets = {
                'mask': to_tensor(transformed['mask']),
                'skeleton': to_tensor(transformed['skeleton']),
                'width': to_tensor(transformed['width']),
                'orientation': to_tensor(transformed['orientation']),
                'junction': to_tensor(transformed['junction']),
                'endpoint': to_tensor(transformed['endpoint']),
            }
            
        else:
            from torchvision import transforms
            image_pil = Image.fromarray(image)
            image_tensor = self.transform(image_pil)
            
            target_h, target_w = self.img_size, self.img_size
            targets = {
                'mask': torch.from_numpy(cv2.resize(mask, (target_w, target_h), interpolation=cv2.INTER_NEAREST)).unsqueeze(0).float(),
                'skeleton': torch.from_numpy(cv2.resize(skeleton, (target_w, target_h), interpolation=cv2.INTER_NEAREST)).unsqueeze(0).float(),
                'width': torch.from_numpy(cv2.resize(width, (target_w, target_h), interpolation=cv2.INTER_LINEAR)).unsqueeze(0).float(),
                'orientation': torch.from_numpy(cv2.resize(orientation, (target_w, target_h), interpolation=cv2.INTER_LINEAR)).unsqueeze(0).float(),
                'junction': torch.from_numpy(cv2.resize(junction, (target_w, target_h), interpolation=cv2.INTER_NEAREST)).unsqueeze(0).float(),
                'endpoint': torch.from_numpy(cv2.resize(endpoint, (target_w, target_h), interpolation=cv2.INTER_NEAREST)).unsqueeze(0).float(),
            }
        
        return image_tensor, targets

# ============================================================
# 5.3 DATASET SPLITTING
# ============================================================

def split_dataset(pairs, train_ratio=0.7, val_ratio=0.15, seed=42):
    """
    Split matched pairs into train/val/test.
    Returns indices.
    """
    np.random.seed(seed)
    n = len(pairs)
    indices = np.random.permutation(n)
    
    n_train = int(n * train_ratio)
    n_val = int(n * val_ratio)
    
    train_idx = indices[:n_train]
    val_idx = indices[n_train:n_train + n_val]
    test_idx = indices[n_train + n_val:]
    
    return train_idx, val_idx, test_idx

# Split RetinoMix-5
rm_pairs = retinomix.pairs
rm_train_idx, rm_val_idx, rm_test_idx = split_dataset(rm_pairs, train_ratio=0.7, val_ratio=0.15, seed=cfg.SEED)

print(f"\n{'='*60}")
print("RETINOMIX-5 SPLIT")
print(f"{'='*60}")
print(f"  Total: {len(rm_pairs)}")
print(f"  Train: {len(rm_train_idx)} ({len(rm_train_idx)/len(rm_pairs)*100:.1f}%)")
print(f"  Val:   {len(rm_val_idx)} ({len(rm_val_idx)/len(rm_pairs)*100:.1f}%)")
print(f"  Test:  {len(rm_test_idx)} ({len(rm_test_idx)/len(rm_pairs)*100:.1f}%)")

# Extract paths
rm_train_images = [rm_pairs[i][0] for i in rm_train_idx]
rm_train_masks = [rm_pairs[i][1] for i in rm_train_idx]
rm_val_images = [rm_pairs[i][0] for i in rm_val_idx]
rm_val_masks = [rm_pairs[i][1] for i in rm_val_idx]
rm_test_images = [rm_pairs[i][0] for i in rm_test_idx]
rm_test_masks = [rm_pairs[i][1] for i in rm_test_idx]

# DRIVE Training (use as-is, or split further)
drive_pairs = drive_train.pairs
drive_train_idx, drive_val_idx, _ = split_dataset(drive_pairs, train_ratio=0.8, val_ratio=0.2, seed=cfg.SEED)

drive_train_images = [drive_pairs[i][0] for i in drive_train_idx]
drive_train_masks = [drive_pairs[i][1] for i in drive_train_idx]
drive_val_images = [drive_pairs[i][0] for i in drive_val_idx]
drive_val_masks = [drive_pairs[i][1] for i in drive_val_idx]

print(f"\n{'='*60}")
print("DRIVE TRAINING SPLIT")
print(f"{'='*60}")
print(f"  Total: {len(drive_pairs)}")
print(f"  Train: {len(drive_train_idx)}")
print(f"  Val:   {len(drive_val_idx)}")

# ============================================================
# 5.4 CREATE DATASETS & DATALOADERS
# ============================================================

print(f"\n{'='*60}")
print("CREATING DATASETS & DATALOADERS")
print(f"{'='*60}")

# Transforms
train_transform, train_transform_type = get_training_augmentation(cfg.IMG_SIZE, cfg.AUG_PROB)
val_transform, val_transform_type = get_validation_augmentation(cfg.IMG_SIZE)

# RetinoMix-5 datasets
rm_train_dataset = RetinalMultiTaskDataset(
    rm_train_images, rm_train_masks,
    precomputed_dir=os.path.join(cfg.PRECOMPUTED_DIR, "retinomix5"),
    transform=train_transform, transform_type=train_transform_type,
    img_size=cfg.IMG_SIZE
)

rm_val_dataset = RetinalMultiTaskDataset(
    rm_val_images, rm_val_masks,
    precomputed_dir=os.path.join(cfg.PRECOMPUTED_DIR, "retinomix5"),
    transform=val_transform, transform_type=val_transform_type,
    img_size=cfg.IMG_SIZE
)

rm_test_dataset = RetinalMultiTaskDataset(
    rm_test_images, rm_test_masks,
    precomputed_dir=os.path.join(cfg.PRECOMPUTED_DIR, "retinomix5"),
    transform=val_transform, transform_type=val_transform_type,
    img_size=cfg.IMG_SIZE
)

# DRIVE datasets
drive_train_dataset = RetinalMultiTaskDataset(
    drive_train_images, drive_train_masks,
    precomputed_dir=os.path.join(cfg.PRECOMPUTED_DIR, "drive_train"),
    transform=train_transform, transform_type=train_transform_type,
    img_size=cfg.IMG_SIZE
)

drive_val_dataset = RetinalMultiTaskDataset(
    drive_val_images, drive_val_masks,
    precomputed_dir=os.path.join(cfg.PRECOMPUTED_DIR, "drive_train"),
    transform=val_transform, transform_type=val_transform_type,
    img_size=cfg.IMG_SIZE
)

# DataLoaders
rm_train_loader = DataLoader(
    rm_train_dataset, batch_size=cfg.BATCH_SIZE, shuffle=True,
    num_workers=cfg.NUM_WORKERS, pin_memory=True, drop_last=True
)
rm_val_loader = DataLoader(
    rm_val_dataset, batch_size=cfg.BATCH_SIZE, shuffle=False,
    num_workers=cfg.NUM_WORKERS, pin_memory=True
)
rm_test_loader = DataLoader(
    rm_test_dataset, batch_size=cfg.BATCH_SIZE, shuffle=False,
    num_workers=cfg.NUM_WORKERS, pin_memory=True
)

drive_train_loader = DataLoader(
    drive_train_dataset, batch_size=cfg.BATCH_SIZE, shuffle=True,
    num_workers=cfg.NUM_WORKERS, pin_memory=True, drop_last=True
)
drive_val_loader = DataLoader(
    drive_val_dataset, batch_size=cfg.BATCH_SIZE, shuffle=False,
    num_workers=cfg.NUM_WORKERS, pin_memory=True
)

print(f"\n{'='*60}")
print("DATALOADER SUMMARY")
print(f"{'='*60}")
print(f"RetinoMix-5 Train: {len(rm_train_dataset)} samples, {len(rm_train_loader)} batches")
print(f"RetinoMix-5 Val:   {len(rm_val_dataset)} samples, {len(rm_val_loader)} batches")
print(f"RetinoMix-5 Test:  {len(rm_test_dataset)} samples, {len(rm_test_loader)} batches")
print(f"DRIVE Train:       {len(drive_train_dataset)} samples, {len(drive_train_loader)} batches")
print(f"DRIVE Val:         {len(drive_val_dataset)} samples, {len(drive_val_loader)} batches")

# ============================================================
# 5.5 VISUALIZE BATCH (inline + saved)
# ============================================================

def visualize_batch(image, targets, save_path=None, max_samples=4):
    """
    Visualize a batch of images and targets.
    Shows inline and saves.
    """
    image = image.cpu().numpy()
    n = min(max_samples, image.shape[0])
    
    fig, axes = plt.subplots(n, 4, figsize=(16, 4*n))
    if n == 1:
        axes = axes.reshape(1, -1)
    
    for i in range(n):
        # Image (denormalize)
        img = image[i].transpose(1, 2, 0)
        img = img * np.array([0.229, 0.224, 0.225]) + np.array([0.485, 0.456, 0.406])
        img = np.clip(img, 0, 1)
        
        axes[i, 0].imshow(img)
        axes[i, 0].set_title("Image", fontsize=10)
        axes[i, 0].axis('off')
        
        # Mask
        axes[i, 1].imshow(targets['mask'][i, 0].cpu().numpy(), cmap='gray')
        axes[i, 1].set_title("Mask", fontsize=10)
        axes[i, 1].axis('off')
        
        # Skeleton
        axes[i, 2].imshow(targets['skeleton'][i, 0].cpu().numpy(), cmap='gray')
        axes[i, 2].set_title("Skeleton", fontsize=10)
        axes[i, 2].axis('off')
        
        # Width
        w = targets['width'][i, 0].cpu().numpy()
        im = axes[i, 3].imshow(w, cmap='plasma')
        axes[i, 3].set_title("Width", fontsize=10)
        axes[i, 3].axis('off')
        plt.colorbar(im, ax=axes[i, 3], fraction=0.046)
    
    plt.suptitle("Training Batch Sample", fontsize=14, fontweight='bold')
    plt.tight_layout()
    
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        print(f"Saved: {save_path}")
    
    plt.show()

# Get one batch and visualize
print(f"\n{'='*60}")
print("VISUALIZING SAMPLE BATCH")
print(f"{'='*60}")

for batch_images, batch_targets in rm_train_loader:
    print(f"Batch image shape: {batch_images.shape}")
    print(f"Batch mask shape: {batch_targets['mask'].shape}")
    print(f"Batch skeleton shape: {batch_targets['skeleton'].shape}")
    
    visualize_batch(
        batch_images, batch_targets,
        save_path=os.path.join(cfg.FIGURES_DIR, "sample_batch.png"),
        max_samples=4
    )
    break  # Only first batch

# Memory check
print(f"\n{'='*60}")
print("MEMORY STATUS")
print(f"{'='*60}")
get_gpu_memory()

print(f"\n{'='*60}")
print("CELL 5 COMPLETE")
print(f"{'='*60}")
```

```python
# ============================================================
# CELL 6: TOPOEYENET ARCHITECTURE (CLEAN — resnet50)
# ============================================================
# Run this AFTER restarting kernel. Do NOT run EfficientNet first.

import torch
import torch.nn as nn
import segmentation_models_pytorch as smp

class TopoEyeNet(nn.Module):
    def __init__(self,
                 encoder_name: str = "resnet50",
                 encoder_weights: str = "imagenet",
                 in_channels: int = 3,
                 mask_head: bool = True,
                 skeleton_head: bool = True):
        super().__init__()
        
        self.mask_head_enabled = mask_head
        self.skeleton_head_enabled = skeleton_head
        
        self.encoder = smp.encoders.get_encoder(
            name=encoder_name,
            in_channels=in_channels,
            depth=5,
            weights=encoder_weights,
        )
        
        encoder_channels = self.encoder.out_channels
        
        self.decoder = smp.decoders.unet.decoder.UnetDecoder(
            encoder_channels=encoder_channels,
            decoder_channels=(256, 128, 64, 32, 16),
            n_blocks=5,
            use_batchnorm=True,
            attention_type=None,
            center=False,
        )
        
        if mask_head:
            self.mask_head = nn.Sequential(
                nn.Conv2d(16, 16, 3, padding=1),
                nn.BatchNorm2d(16),
                nn.ReLU(inplace=True),
                nn.Conv2d(16, 1, 1),
            )
        if skeleton_head:
            self.skeleton_head = nn.Sequential(
                nn.Conv2d(16, 16, 3, padding=1),
                nn.BatchNorm2d(16),
                nn.ReLU(inplace=True),
                nn.Conv2d(16, 1, 1),
            )
        
        self._init_weights()
    
    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Conv2d):
                nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
    
    def forward(self, x):
        features = self.encoder(x)
        decoder_output = self.decoder(*features)
        
        outputs = {}
        if self.mask_head_enabled:
            outputs['mask'] = self.mask_head(decoder_output)
        if self.skeleton_head_enabled:
            outputs['skeleton'] = self.skeleton_head(decoder_output)
        
        return outputs

# Build and test
print("=" * 60)
print("BUILDING TOPOEYENET")
print("=" * 60)

model = TopoEyeNet(
    encoder_name="resnet50",
    encoder_weights="imagenet",
    in_channels=3,
    mask_head=True,
    skeleton_head=True,
)

total_params = sum(p.numel() for p in model.parameters())
print(f"Total parameters: {total_params:,}")

# Move to GPU
model = model.to(device)
print(f"Model on: {next(model.parameters()).device}")

# Test forward
print("\nTesting forward pass...")
test_input = torch.randn(2, 3, cfg.IMG_SIZE, cfg.IMG_SIZE).to(device)
with torch.no_grad():
    test_output = model(test_input)

print(f"Input: {test_input.shape}")
for k, v in test_output.items():
    print(f"  {k}: {v.shape}")

# Wrap for dual GPU AFTER confirming single GPU works
if torch.cuda.device_count() > 1:
    print(f"\nWrapping for {torch.cuda.device_count()} GPUs...")
    model = nn.DataParallel(model, device_ids=[0, 1])
    
    # Re-test with DataParallel
    test_input = torch.randn(4, 3, cfg.IMG_SIZE, cfg.IMG_SIZE).to(device)
    with torch.no_grad():
        test_output = model(test_input)
    print(f"DataParallel test — Input: {test_input.shape}")
    for k, v in test_output.items():
        print(f"  {k}: {v.shape}")

get_gpu_memory()
print("\nCELL 6 COMPLETE")
```

```python
# ============================================================
# CELL 7: LOSS FUNCTIONS
# ============================================================
# 1. Dice Loss (for mask + skeleton segmentation)
# 2. Structural Consistency Loss (novelty — mask/skeleton agreement)
# 3. Combined multi-task loss with weighting

import torch
import torch.nn as nn
import torch.nn.functional as F

# ============================================================
# 7.1 DICE LOSS
# ============================================================

class DiceLoss(nn.Module):
    """
    Dice loss for binary segmentation.
    Handles class imbalance better than BCE.
    """
    def __init__(self, smooth: float = 1.0, eps: float = 1e-7):
        super().__init__()
        self.smooth = smooth
        self.eps = eps
    
    def forward(self, pred: torch.Tensor, target: torch.Tensor) -> torch.Tensor:
        """
        Args:
            pred: (B, 1, H, W) logits or probabilities
            target: (B, 1, H, W) binary ground truth
        """
        # Apply sigmoid if logits
        pred = torch.sigmoid(pred)
        
        # Flatten
        pred = pred.view(pred.size(0), -1)
        target = target.view(target.size(0), -1)
        
        intersection = (pred * target).sum(dim=1)
        union = pred.sum(dim=1) + target.sum(dim=1)
        
        dice = (2.0 * intersection + self.smooth) / (union + self.smooth + self.eps)
        return 1.0 - dice.mean()

class DiceBCELoss(nn.Module):
    """
    Combined Dice + BCE loss for robust segmentation.
    """
    def __init__(self, dice_weight: float = 0.5, bce_weight: float = 0.5):
        super().__init__()
        self.dice_weight = dice_weight
        self.bce_weight = bce_weight
        self.dice = DiceLoss()
        self.bce = nn.BCEWithLogitsLoss()
    
    def forward(self, pred: torch.Tensor, target: torch.Tensor) -> torch.Tensor:
        dice_loss = self.dice(pred, target)
        bce_loss = self.bce(pred, target)
        return self.dice_weight * dice_loss + self.bce_weight * bce_loss

# ============================================================
# 7.2 STRUCTURAL CONSISTENCY LOSS (NOVELTY)
# ============================================================

class StructuralConsistencyLoss(nn.Module):
    """
    Enforces anatomical agreement between predicted mask and skeleton.
    
    Core idea: A pixel predicted as 'skeleton' MUST also be predicted as 'vessel' in the mask.
    A pixel predicted as 'background' in the mask CANNOT have a skeleton pixel.
    
    This is stronger than just adding another auxiliary decoder — it creates
    a logical constraint between representations.
    """
    
    def __init__(self, mode: str = "soft", eps: float = 1e-7):
        """
        Args:
            mode: 'soft' (differentiable) or 'hard' (binary, for eval only)
            eps: numerical stability
        """
        super().__init__()
        self.mode = mode
        self.eps = eps
    
    def forward(self, mask_pred: torch.Tensor, skeleton_pred: torch.Tensor) -> torch.Tensor:
        """
        Args:
            mask_pred: (B, 1, H, W) mask logits
            skeleton_pred: (B, 1, H, W) skeleton logits
        
        Returns:
            consistency_loss: scalar tensor
        """
        # Convert to probabilities
        mask_prob = torch.sigmoid(mask_pred)
        skeleton_prob = torch.sigmoid(skeleton_pred)
        
        if self.mode == "soft":
            # Soft consistency: skeleton_prob should be <= mask_prob
            # Violation: skeleton_prob > mask_prob
            violation = F.relu(skeleton_prob - mask_prob)  # positive where skeleton > mask
            
            # Weight by skeleton confidence: high-confidence violations are worse
            weighted_violation = violation * skeleton_prob
            
            # Normalize by number of skeleton pixels (approximate)
            consistency_loss = weighted_violation.sum() / (skeleton_prob.sum() + self.eps)
            
        elif self.mode == "hard":
            # Hard consistency: binary check (non-differentiable, for analysis)
            mask_bin = (mask_prob > 0.5).float()
            skeleton_bin = (skeleton_prob > 0.5).float()
            
            # Impossible pixels: skeleton=1 but mask=0
            impossible = skeleton_bin * (1 - mask_bin)
            consistency_loss = impossible.sum() / (skeleton_bin.sum() + self.eps)
        
        return consistency_loss

class MaskSkeletonAgreementLoss(nn.Module):
    """
    Extended consistency: skeleton should be a subset of mask,
    AND mask should contain the skeleton (coverage loss).
    """
    
    def __init__(self, eps: float = 1e-7):
        super().__init__()
        self.eps = eps
        self.consistency = StructuralConsistencyLoss(mode="soft")
    
    def forward(self, mask_pred: torch.Tensor, skeleton_pred: torch.Tensor) -> torch.Tensor:
        # 1. Skeleton must be subset of mask (consistency)
        subset_loss = self.consistency(mask_pred, skeleton_pred)
        
        # 2. Mask should cover skeleton (coverage)
        # Where skeleton is confident, mask should also be confident
        mask_prob = torch.sigmoid(mask_pred)
        skeleton_prob = torch.sigmoid(skeleton_pred)
        
        # Coverage: mask_prob should be high where skeleton_prob is high
        coverage = skeleton_prob * (1 - mask_prob)  # high when skeleton confident but mask not
        coverage_loss = coverage.sum() / (skeleton_prob.sum() + self.eps)
        
        # 3. Skeleton should be thin (sparsity regularization)
        # Encourage skeleton to be sparse (fewer pixels than mask)
        sparsity_loss = skeleton_prob.mean() / (mask_prob.mean() + self.eps)
        sparsity_loss = F.relu(sparsity_loss - 0.3)  # only penalize if > 30% of mask
        
        return subset_loss + coverage_loss + 0.1 * sparsity_loss

# ============================================================
# 7.3 COMBINED MULTI-TASK LOSS
# ============================================================

class TopoEyeLoss(nn.Module):
    """
    Combined loss for TopoEyeNet multi-task learning.
    
    Loss = λ_mask * DiceBCE(mask) 
         + λ_skeleton * DiceBCE(skeleton)
         + λ_consistency * StructuralConsistency(mask, skeleton)
    """
    
    def __init__(self,
                 lambda_mask: float = 1.0,
                 lambda_skeleton: float = 0.5,
                 lambda_consistency: float = 0.3,
                 lambda_width: float = 0.0,
                 lambda_orientation: float = 0.0,
                 lambda_junction: float = 0.0):
        super().__init__()
        
        self.lambda_mask = lambda_mask
        self.lambda_skeleton = lambda_skeleton
        self.lambda_consistency = lambda_consistency
        self.lambda_width = lambda_width
        self.lambda_orientation = lambda_orientation
        self.lambda_junction = lambda_junction
        
        self.mask_loss = DiceBCELoss(dice_weight=0.7, bce_weight=0.3)
        self.skeleton_loss = DiceBCELoss(dice_weight=0.7, bce_weight=0.3)
        self.consistency_loss = MaskSkeletonAgreementLoss()
        
        # v2+ losses
        self.width_loss = nn.L1Loss() if lambda_width > 0 else None
        self.orientation_loss = nn.L1Loss() if lambda_orientation > 0 else None
        self.junction_loss = DiceBCELoss() if lambda_junction > 0 else None
    
    def forward(self, predictions: dict, targets: dict) -> dict:
        """
        Args:
            predictions: dict from model forward {'mask': ..., 'skeleton': ...}
            targets: dict from dataloader {'mask': ..., 'skeleton': ...}
        
        Returns:
            dict with 'total' loss and individual components
        """
        losses = {}
        total_loss = 0.0
        
        # Mask loss
        if 'mask' in predictions and 'mask' in targets:
            losses['mask'] = self.mask_loss(predictions['mask'], targets['mask'])
            total_loss += self.lambda_mask * losses['mask']
        
        # Skeleton loss
        if 'skeleton' in predictions and 'skeleton' in targets:
            losses['skeleton'] = self.skeleton_loss(predictions['skeleton'], targets['skeleton'])
            total_loss += self.lambda_skeleton * losses['skeleton']
        
        # Structural consistency (novelty)
        if 'mask' in predictions and 'skeleton' in predictions:
            losses['consistency'] = self.consistency_loss(predictions['mask'], predictions['skeleton'])
            total_loss += self.lambda_consistency * losses['consistency']
        
        # Width loss (v2)
        if self.lambda_width > 0 and 'width' in predictions and 'width' in targets:
            losses['width'] = self.width_loss(
                torch.sigmoid(predictions['width']) * targets['mask'],  # only where vessel
                targets['width'] * targets['mask']
            )
            total_loss += self.lambda_width * losses['width']
        
        # Orientation loss (v2)
        if self.lambda_orientation > 0 and 'orientation' in predictions and 'orientation' in targets:
            losses['orientation'] = self.orientation_loss(
                predictions['orientation'] * targets['mask'],
                targets['orientation'] * targets['mask']
            )
            total_loss += self.lambda_orientation * losses['orientation']
        
        # Junction loss (v2)
        if self.lambda_junction > 0 and 'junction' in predictions and 'junction' in targets:
            losses['junction'] = self.junction_loss(predictions['junction'], targets['junction'])
            total_loss += self.lambda_junction * losses['junction']
        
        losses['total'] = total_loss
        return losses

# ============================================================
# 7.4 INSTANTIATE & TEST
# ============================================================

print("=" * 60)
print("BUILDING LOSS FUNCTIONS")
print("=" * 60)

criterion = TopoEyeLoss(
    lambda_mask=cfg.LAMBDA_MASK,
    lambda_skeleton=cfg.LAMBDA_SKELETON,
    lambda_consistency=cfg.LAMBDA_CONSISTENCY,
)

print(f"Loss weights:")
print(f"  λ_mask: {cfg.LAMBDA_MASK}")
print(f"  λ_skeleton: {cfg.LAMBDA_SKELETON}")
print(f"  λ_consistency: {cfg.LAMBDA_CONSISTENCY}")

# Test with dummy data
print(f"\nTesting loss computation...")
model.eval()
with torch.no_grad():
    dummy_pred = model(test_input)

# Create dummy targets
dummy_targets = {
    'mask': torch.randint(0, 2, (4, 1, 512, 512)).float().to(device),
    'skeleton': torch.randint(0, 2, (4, 1, 512, 512)).float().to(device),
}

losses = criterion(dummy_pred, dummy_targets)
print(f"\nLoss components:")
for k, v in losses.items():
    print(f"  {k}: {v.item():.4f}")

# Visualize consistency concept
print(f"\n{'='*60}")
print("STRUCTURAL CONSISTENCY VISUALIZATION")
print(f"{'='*60}")

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Simulate a case where skeleton extends beyond mask (violation)
mask = torch.zeros(1, 1, 128, 128)
mask[:, :, 30:100, 30:100] = 1.0

skeleton_good = torch.zeros(1, 1, 128, 128)
skeleton_good[:, :, 50:80, 50:80] = 1.0  # Inside mask

skeleton_bad = torch.zeros(1, 1, 128, 128)
skeleton_bad[:, :, 50:80, 50:80] = 1.0
skeleton_bad[:, :, 20:40, 20:40] = 1.0  # Outside mask (violation!)

axes[0].imshow(mask[0, 0].cpu(), cmap='gray')
axes[0].set_title("Mask")
axes[0].axis('off')

axes[1].imshow(skeleton_good[0, 0].cpu(), cmap='gray')
axes[1].set_title("Skeleton (valid)\nSubset of mask")
axes[1].axis('off')

axes[2].imshow(skeleton_bad[0, 0].cpu(), cmap='gray')
axes[2].set_title("Skeleton (invalid)\nExtends beyond mask")
axes[2].axis('off')

plt.suptitle("Structural Consistency Concept", fontsize=14, fontweight='bold')
plt.tight_layout()
save_path = os.path.join(cfg.FIGURES_DIR, "consistency_concept.png")
plt.savefig(save_path, dpi=150, bbox_inches='tight')
print(f"Saved: {save_path}")
plt.show()

get_gpu_memory()
print(f"\n{'='*60}")
print("CELL 7 COMPLETE")
print(f"{'='*60}")
```

```python
# ============================================================
# CELL 8: TRAINING LOOP
# ============================================================
# Full training with AMP, gradient accumulation, checkpointing,
# validation, early stopping, and metric logging.

import os
import time
import json
from collections import defaultdict
from torch.cuda.amp import autocast, GradScaler

# ============================================================
# 8.1 METRICS
# ============================================================

class SegmentationMetrics:
    """Compute Dice, IoU, Precision, Recall, F1 for binary segmentation."""
    
    @staticmethod
    def dice_score(pred: torch.Tensor, target: torch.Tensor, threshold: float = 0.5) -> float:
        pred_bin = (torch.sigmoid(pred) > threshold).float()
        target_bin = target.float()
        
        intersection = (pred_bin * target_bin).sum()
        union = pred_bin.sum() + target_bin.sum()
        
        return (2.0 * intersection / (union + 1e-7)).item()
    
    @staticmethod
    def iou_score(pred: torch.Tensor, target: torch.Tensor, threshold: float = 0.5) -> float:
        pred_bin = (torch.sigmoid(pred) > threshold).float()
        target_bin = target.float()
        
        intersection = (pred_bin * target_bin).sum()
        union = pred_bin.sum() + target_bin.sum() - intersection
        
        return (intersection / (union + 1e-7)).item()
    
    @staticmethod
    def compute_all(pred: torch.Tensor, target: torch.Tensor, threshold: float = 0.5) -> dict:
        pred_prob = torch.sigmoid(pred)
        pred_bin = (pred_prob > threshold).float()
        target_bin = target.float()
        
        tp = (pred_bin * target_bin).sum().item()
        fp = (pred_bin * (1 - target_bin)).sum().item()
        fn = ((1 - pred_bin) * target_bin).sum().item()
        tn = ((1 - pred_bin) * (1 - target_bin)).sum().item()
        
        dice = 2 * tp / (2 * tp + fp + fn + 1e-7)
        iou = tp / (tp + fp + fn + 1e-7)
        precision = tp / (tp + fp + 1e-7)
        recall = tp / (tp + fn + 1e-7)
        f1 = 2 * precision * recall / (precision + recall + 1e-7)
        
        return {
            'dice': dice,
            'iou': iou,
            'precision': precision,
            'recall': recall,
            'f1': f1,
        }

# ============================================================
# 8.2 TRAINER CLASS
# ============================================================

class Trainer:
    def __init__(self,
                 model,
                 criterion,
                 optimizer,
                 scheduler,
                 train_loader,
                 val_loader,
                 device,
                 cfg,
                 scaler=None):
        
        self.model = model
        self.criterion = criterion
        self.optimizer = optimizer
        self.scheduler = scheduler
        self.train_loader = train_loader
        self.val_loader = val_loader
        self.device = device
        self.cfg = cfg
        self.scaler = scaler
        
        self.best_val_dice = 0.0
        self.epochs_no_improve = 0
        self.history = defaultdict(list)
        
        # Create checkpoint dir
        os.makedirs(cfg.CHECKPOINT_DIR, exist_ok=True)
        
    def train_epoch(self, epoch: int):
        self.model.train()
        epoch_losses = defaultdict(float)
        epoch_metrics = defaultdict(float)
        num_batches = 0
        
        pbar = self.train_loader
        for batch_idx, (images, targets) in enumerate(pbar):
            images = images.to(self.device, non_blocking=True)
            targets = {k: v.to(self.device, non_blocking=True) for k, v in targets.items()}
            
            # Forward with AMP
            with autocast(enabled=self.cfg.USE_AMP):
                outputs = self.model(images)
                losses = self.criterion(outputs, targets)
                loss = losses['total'] / self.cfg.GRADIENT_ACCUMULATION_STEPS
            
            # Backward with AMP
            if self.cfg.USE_AMP and self.scaler:
                self.scaler.scale(loss).backward()
            else:
                loss.backward()
            
            # Gradient accumulation
            if (batch_idx + 1) % self.cfg.GRADIENT_ACCUMULATION_STEPS == 0:
                if self.cfg.USE_AMP and self.scaler:
                    self.scaler.step(self.optimizer)
                    self.scaler.update()
                else:
                    self.optimizer.step()
                self.optimizer.zero_grad()
            
            # Accumulate metrics
            for k, v in losses.items():
                epoch_losses[k] += v.item()
            
            # Compute batch metrics (on mask only for speed)
            with torch.no_grad():
                mask_metrics = SegmentationMetrics.compute_all(outputs['mask'], targets['mask'])
                for k, v in mask_metrics.items():
                    epoch_metrics[f'mask_{k}'] += v
            
            num_batches += 1
            
            # Clear cache periodically
            if (batch_idx + 1) % self.cfg.CLEAR_CACHE_EVERY_N_BATCHES == 0:
                clear_gpu_cache()
            
            # Log interval
            if (batch_idx + 1) % self.cfg.LOG_INTERVAL == 0:
                print(f"  Epoch {epoch} | Batch {batch_idx+1}/{len(self.train_loader)} | "
                      f"Loss: {losses['total'].item():.4f} | "
                      f"Mask Dice: {mask_metrics['dice']:.4f}")
        
        # Average
        for k in epoch_losses:
            epoch_losses[k] /= num_batches
        for k in epoch_metrics:
            epoch_metrics[k] /= num_batches
        
        return epoch_losses, epoch_metrics
    
    @torch.no_grad()
    def validate(self, epoch: int):
        self.model.eval()
        val_losses = defaultdict(float)
        val_metrics = defaultdict(float)
        num_batches = 0
        
        for images, targets in self.val_loader:
            images = images.to(self.device, non_blocking=True)
            targets = {k: v.to(self.device, non_blocking=True) for k, v in targets.items()}
            
            with autocast(enabled=self.cfg.USE_AMP):
                outputs = self.model(images)
                losses = self.criterion(outputs, targets)
            
            for k, v in losses.items():
                val_losses[k] += v.item()
            
            # Metrics
            mask_metrics = SegmentationMetrics.compute_all(outputs['mask'], targets['mask'])
            skel_metrics = SegmentationMetrics.compute_all(outputs['skeleton'], targets['skeleton'])
            
            for k, v in mask_metrics.items():
                val_metrics[f'mask_{k}'] += v
            for k, v in skel_metrics.items():
                val_metrics[f'skeleton_{k}'] += v
            
            num_batches += 1
        
        # Average
        for k in val_losses:
            val_losses[k] /= num_batches
        for k in val_metrics:
            val_metrics[k] /= num_batches
        
        return val_losses, val_metrics
    
    def save_checkpoint(self, epoch: int, is_best: bool = False):
        checkpoint = {
            'epoch': epoch,
            'model_state_dict': self.model.module.state_dict() if hasattr(self.model, 'module') else self.model.state_dict(),
            'optimizer_state_dict': self.optimizer.state_dict(),
            'scheduler_state_dict': self.scheduler.state_dict(),
            'best_val_dice': self.best_val_dice,
            'history': dict(self.history),
        }
        
        path = os.path.join(self.cfg.CHECKPOINT_DIR, f"{self.cfg.EXPERIMENT_NAME}_epoch{epoch}.pt")
        torch.save(checkpoint, path)
        print(f"  Saved checkpoint: {path}")
        
        if is_best:
            best_path = os.path.join(self.cfg.CHECKPOINT_DIR, f"{self.cfg.EXPERIMENT_NAME}_best.pt")
            torch.save(checkpoint, best_path)
            print(f"  Saved BEST checkpoint: {best_path}")
    
    def fit(self, epochs: int):
        print(f"\n{'='*60}")
        print("STARTING TRAINING")
        print(f"{'='*60}")
        print(f"Epochs: {epochs}")
        print(f"Train batches: {len(self.train_loader)}")
        print(f"Val batches: {len(self.val_loader)}")
        print(f"AMP: {self.cfg.USE_AMP}")
        print(f"Gradient Accumulation: {self.cfg.GRADIENT_ACCUMULATION_STEPS}")
        
        for epoch in range(1, epochs + 1):
            start_time = time.time()
            
            # Train
            train_losses, train_metrics = self.train_epoch(epoch)
            
            # Validate
            if epoch % self.cfg.VAL_FREQ == 0:
                val_losses, val_metrics = self.validate(epoch)
                
                # Scheduler step
                if isinstance(self.scheduler, torch.optim.lr_scheduler.ReduceLROnPlateau):
                    self.scheduler.step(val_losses['total'])
                else:
                    self.scheduler.step()
                
                # Log
                print(f"\n{'='*60}")
                print(f"EPOCH {epoch}/{epochs} | Time: {time.time()-start_time:.1f}s")
                print(f"{'='*60}")
                print(f"Train Loss: {train_losses['total']:.4f} | "
                      f"Mask Dice: {train_metrics['mask_dice']:.4f}")
                print(f"Val Loss:   {val_losses['total']:.4f} | "
                      f"Mask Dice: {val_metrics['mask_dice']:.4f} | "
                      f"Skeleton Dice: {val_metrics['skeleton_dice']:.4f}")
                
                # History
                self.history['epoch'].append(epoch)
                self.history['train_loss'].append(train_losses['total'])
                self.history['val_loss'].append(val_losses['total'])
                self.history['val_mask_dice'].append(val_metrics['mask_dice'])
                self.history['val_skeleton_dice'].append(val_metrics['skeleton_dice'])
                
                # Early stopping check
                current_dice = val_metrics['mask_dice']
                if current_dice > self.best_val_dice:
                    self.best_val_dice = current_dice
                    self.epochs_no_improve = 0
                    if self.cfg.SAVE_BEST_ONLY:
                        self.save_checkpoint(epoch, is_best=True)
                else:
                    self.epochs_no_improve += 1
                
                if self.epochs_no_improve >= self.cfg.EARLY_STOPPING_PATIENCE:
                    print(f"\nEarly stopping triggered after {epoch} epochs")
                    break
                
                # Regular checkpoint
                if not self.cfg.SAVE_BEST_ONLY:
                    self.save_checkpoint(epoch)
            
            else:
                # No validation this epoch
                print(f"\nEpoch {epoch} | Train Loss: {train_losses['total']:.4f} | "
                      f"Mask Dice: {train_metrics['mask_dice']:.4f}")
                self.scheduler.step()
        
        # Save final
        self.save_checkpoint(epochs)
        
        # Save history
        history_path = os.path.join(self.cfg.LOG_DIR, 'training_history.json')
        with open(history_path, 'w') as f:
            json.dump(dict(self.history), f, indent=2)
        print(f"\nSaved history: {history_path}")
        
        return self.history

# ============================================================
# 8.3 OPTIMIZER & SCHEDULER
# ============================================================

print("=" * 60)
print("SETTING UP OPTIMIZER & SCHEDULER")
print("=" * 60)

# Optimizer
if cfg.OPTIMIZER.lower() == "adamw":
    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=cfg.LEARNING_RATE,
        weight_decay=cfg.WEIGHT_DECAY,
    )
elif cfg.OPTIMIZER.lower() == "adam":
    optimizer = torch.optim.Adam(
        model.parameters(),
        lr=cfg.LEARNING_RATE,
        weight_decay=cfg.WEIGHT_DECAY,
    )
else:
    optimizer = torch.optim.SGD(
        model.parameters(),
        lr=cfg.LEARNING_RATE,
        momentum=0.9,
        weight_decay=cfg.WEIGHT_DECAY,
    )

# Scheduler
if cfg.SCHEDULER.lower() == "cosine":
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
        optimizer, T_max=cfg.T_MAX, eta_min=cfg.ETA_MIN
    )
elif cfg.SCHEDULER.lower() == "plateau":
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode='min', patience=5, factor=0.5
    )
else:
    scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=30, gamma=0.1)

# AMP Scaler
scaler = GradScaler() if cfg.USE_AMP else None

print(f"Optimizer: {cfg.OPTIMIZER}")
print(f"Scheduler: {cfg.SCHEDULER}")
print(f"Learning rate: {cfg.LEARNING_RATE}")
print(f"AMP: {cfg.USE_AMP}")

# ============================================================
# 8.4 TRAIN
# ============================================================

trainer = Trainer(
    model=model,
    criterion=criterion,
    optimizer=optimizer,
    scheduler=scheduler,
    train_loader=rm_train_loader,
    val_loader=rm_val_loader,
    device=device,
    cfg=cfg,
    scaler=scaler,
)

# Quick sanity check: run 1 batch to verify everything works
print(f"\n{'='*60}")
print("SANITY CHECK: 1 batch")
print(f"{'='*60}")

model.train()
for images, targets in rm_train_loader:
    images = images.to(device)
    targets = {k: v.to(device) for k, v in targets.items()}
    
    with autocast(enabled=cfg.USE_AMP):
        outputs = model(images)
        losses = criterion(outputs, targets)
    
    print(f"Forward pass OK")
    print(f"  Loss components: { {k: f'{v.item():.4f}' for k, v in losses.items()} }")
    
    # Backward
    if cfg.USE_AMP and scaler:
        scaler.scale(losses['total']).backward()
        scaler.step(optimizer)
        scaler.update()
    else:
        losses['total'].backward()
        optimizer.step()
    
    optimizer.zero_grad()
    print(f"  Backward pass OK")
    break

clear_gpu_cache()

# Full training (reduce epochs for testing)
print(f"\n{'='*60}")
print("FULL TRAINING")
print(f"{'='*60}")

# For quick testing, use fewer epochs
TRAIN_EPOCHS = 5  # Change to cfg.EPOCHS for full training

history = trainer.fit(epochs=TRAIN_EPOCHS)

# Plot training curves
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(history['epoch'], history['train_loss'], label='Train Loss', marker='o')
axes[0].plot(history['epoch'], history['val_loss'], label='Val Loss', marker='o')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].set_title('Training & Validation Loss')
axes[0].legend()
axes[0].grid(True)

axes[1].plot(history['epoch'], history['val_mask_dice'], label='Mask Dice', marker='o')
axes[1].plot(history['epoch'], history['val_skeleton_dice'], label='Skeleton Dice', marker='o')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Dice Score')
axes[1].set_title('Validation Dice Scores')
axes[1].legend()
axes[1].grid(True)

plt.suptitle("TopoEyeNet Training Curves", fontsize=14, fontweight='bold')
plt.tight_layout()
save_path = os.path.join(cfg.FIGURES_DIR, "training_curves.png")
plt.savefig(save_path, dpi=150, bbox_inches='tight')
print(f"\nSaved: {save_path}")
plt.show()

get_gpu_memory()
print(f"\n{'='*60}")
print("CELL 8 COMPLETE")
print(f"{'='*60}")
```

```python
# ============================================================
# CELL 9: IMPROVED LOSS & TRAINING STRATEGY
# ============================================================
# Fixes for sparse target learning:
#   1. Weighted BCE + Focal Loss for skeleton (sparse pixels get higher weight)
#   2. Tversky Loss (controls precision/recall tradeoff for imbalanced data)
#   3. Consistency loss curriculum (ramps up over epochs)
#   4. Cosine warmup LR scheduler
#   5. Deep supervision (auxiliary losses at multiple decoder levels)

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.cuda.amp import autocast, GradScaler
import numpy as np

# ============================================================
# 9.1 IMPROVED LOSS FUNCTIONS
# ============================================================

class FocalLoss(nn.Module):
    """
    Focal Loss for extremely imbalanced segmentation.
    Down-weights easy examples, focuses on hard examples.
    """
    def __init__(self, alpha: float = 0.25, gamma: float = 2.0, reduction: str = 'mean'):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.reduction = reduction
    
    def forward(self, pred: torch.Tensor, target: torch.Tensor) -> torch.Tensor:
        bce = F.binary_cross_entropy_with_logits(pred, target, reduction='none')
        pred_prob = torch.sigmoid(pred)
        p_t = pred_prob * target + (1 - pred_prob) * (1 - target)
        focal_weight = self.alpha * (1 - p_t) ** self.gamma
        loss = focal_weight * bce
        
        if self.reduction == 'mean':
            return loss.mean()
        elif self.reduction == 'sum':
            return loss.sum()
        return loss

class TverskyLoss(nn.Module):
    """
    Tversky loss: generalization of Dice that controls false positive/negative tradeoff.
    beta > 0.5 penalizes false negatives more (good for sparse targets like skeleton).
    """
    def __init__(self, alpha: float = 0.3, beta: float = 0.7, smooth: float = 1.0):
        super().__init__()
        self.alpha = alpha  # weight for false positives
        self.beta = beta    # weight for false negatives (higher = recall-focused)
        self.smooth = smooth
    
    def forward(self, pred: torch.Tensor, target: torch.Tensor) -> torch.Tensor:
        pred = torch.sigmoid(pred)
        pred = pred.view(pred.size(0), -1)
        target = target.view(target.size(0), -1)
        
        tp = (pred * target).sum(dim=1)
        fp = (pred * (1 - target)).sum(dim=1)
        fn = ((1 - pred) * target).sum(dim=1)
        
        tversky = (tp + self.smooth) / (tp + self.alpha * fp + self.beta * fn + self.smooth)
        return 1 - tversky.mean()

class WeightedDiceLoss(nn.Module):
    """
    Dice loss with per-pixel weighting.
    Rare classes (skeleton) get higher weight.
    """
    def __init__(self, pos_weight: float = 10.0, smooth: float = 1.0):
        super().__init__()
        self.pos_weight = pos_weight
        self.smooth = smooth
    
    def forward(self, pred: torch.Tensor, target: torch.Tensor) -> torch.Tensor:
        pred = torch.sigmoid(pred)
        pred = pred.view(pred.size(0), -1)
        target = target.view(target.size(0), -1)
        
        # Weight positive pixels higher
        weights = torch.ones_like(target)
        weights[target > 0.5] = self.pos_weight
        
        intersection = (pred * target * weights).sum(dim=1)
        union = (pred * weights).sum(dim=1) + (target * weights).sum(dim=1)
        
        dice = (2.0 * intersection + self.smooth) / (union + self.smooth)
        return 1 - dice.mean()

class ConsistencyLossCurriculum(nn.Module):
    """
    Structural consistency loss with curriculum:
    - Starts at 0, ramps up to full weight over warmup_epochs
    - Allows mask and skeleton to learn independently first
    """
    def __init__(self, warmup_epochs: int = 10, full_weight: float = 0.3):
        super().__init__()
        self.warmup_epochs = warmup_epochs
        self.full_weight = full_weight
        self.base_consistency = MaskSkeletonAgreementLoss()
    
    def get_weight(self, epoch: int) -> float:
        if epoch <= 0:
            return 0.0
        if epoch >= self.warmup_epochs:
            return self.full_weight
        # Linear ramp
        return self.full_weight * (epoch / self.warmup_epochs)
    
    def forward(self, mask_pred: torch.Tensor, skeleton_pred: torch.Tensor, epoch: int = 0) -> torch.Tensor:
        weight = self.get_weight(epoch)
        if weight == 0:
            return torch.tensor(0.0, device=mask_pred.device)
        return weight * self.base_consistency(mask_pred, skeleton_pred)

# ============================================================
# 9.2 IMPROVED COMBINED LOSS
# ============================================================

class TopoEyeLossV2(nn.Module):
    """
    Improved multi-task loss with:
    - Tversky loss for mask (balanced precision/recall)
    - Focal + Weighted Dice for skeleton (sparse target)
    - Curriculum consistency loss
    """
    
    def __init__(self,
                 lambda_mask: float = 1.0,
                 lambda_skeleton: float = 1.0,
                 lambda_consistency: float = 0.3,
                 consistency_warmup: int = 10):
        super().__init__()
        
        self.lambda_mask = lambda_mask
        self.lambda_skeleton = lambda_skeleton
        self.lambda_consistency = lambda_consistency
        
        self.mask_loss = TverskyLoss(alpha=0.3, beta=0.7)  # Slight recall bias for vessels
        self.skeleton_loss = nn.ModuleDict({
            'focal': FocalLoss(alpha=0.25, gamma=2.0),
            'dice': WeightedDiceLoss(pos_weight=50.0),  # Heavy weight on skeleton pixels
        })
        self.consistency = ConsistencyLossCurriculum(
            warmup_epochs=consistency_warmup,
            full_weight=lambda_consistency
        )
    
    def forward(self, predictions: dict, targets: dict, epoch: int = 0) -> dict:
        losses = {}
        total_loss = 0.0
        
        # Mask loss
        if 'mask' in predictions and 'mask' in targets:
            losses['mask'] = self.mask_loss(predictions['mask'], targets['mask'])
            total_loss += self.lambda_mask * losses['mask']
        
        # Skeleton loss: combine focal + weighted dice
        if 'skeleton' in predictions and 'skeleton' in targets:
            focal_loss = self.skeleton_loss['focal'](predictions['skeleton'], targets['skeleton'])
            dice_loss = self.skeleton_loss['dice'](predictions['skeleton'], targets['skeleton'])
            losses['skeleton_focal'] = focal_loss
            losses['skeleton_dice'] = dice_loss
            losses['skeleton'] = 0.5 * focal_loss + 0.5 * dice_loss
            total_loss += self.lambda_skeleton * losses['skeleton']
        
        # Consistency with curriculum
        if 'mask' in predictions and 'skeleton' in predictions:
            losses['consistency'] = self.consistency(predictions['mask'], predictions['skeleton'], epoch)
            total_loss += losses['consistency']  # Already weighted inside curriculum
        
        losses['total'] = total_loss
        return losses

# ============================================================
# 9.3 LR SCHEDULER WITH WARMUP
# ============================================================

class CosineWarmupScheduler:
    """
    Cosine annealing with linear warmup.
    """
    def __init__(self, optimizer, warmup_epochs: int, max_epochs: int, 
                 base_lr: float, max_lr: float, min_lr: float = 1e-6):
        self.optimizer = optimizer
        self.warmup_epochs = warmup_epochs
        self.max_epochs = max_epochs
        self.base_lr = base_lr
        self.max_lr = max_lr
        self.min_lr = min_lr
        self.current_epoch = 0
        
        # Set initial LR
        for param_group in self.optimizer.param_groups:
            param_group['lr'] = base_lr
    
    def step(self, epoch: int = None):
        if epoch is not None:
            self.current_epoch = epoch
        else:
            self.current_epoch += 1
        
        e = self.current_epoch
        
        if e < self.warmup_epochs:
            # Linear warmup
            lr = self.base_lr + (self.max_lr - self.base_lr) * (e / self.warmup_epochs)
        else:
            # Cosine annealing
            progress = (e - self.warmup_epochs) / (self.max_epochs - self.warmup_epochs)
            lr = self.min_lr + (self.max_lr - self.min_lr) * 0.5 * (1 + np.cos(np.pi * progress))
        
        for param_group in self.optimizer.param_groups:
            param_group['lr'] = lr
        
        return lr
    
    def state_dict(self):
        return {'current_epoch': self.current_epoch}
    
    def load_state_dict(self, state_dict):
        self.current_epoch = state_dict['current_epoch']

# ============================================================
# 9.4 UPDATED TRAINER WITH IMPROVEMENTS
# ============================================================

class TrainerV2:
    def __init__(self, model, criterion, optimizer, scheduler, 
                 train_loader, val_loader, device, cfg, scaler=None):
        self.model = model
        self.criterion = criterion
        self.optimizer = optimizer
        self.scheduler = scheduler
        self.train_loader = train_loader
        self.val_loader = val_loader
        self.device = device
        self.cfg = cfg
        self.scaler = scaler
        
        self.best_val_dice = 0.0
        self.epochs_no_improve = 0
        self.history = {k: [] for k in ['epoch', 'train_loss', 'val_loss', 
                                         'val_mask_dice', 'val_skeleton_dice', 'lr']}
        
        os.makedirs(cfg.CHECKPOINT_DIR, exist_ok=True)
    
    def train_epoch(self, epoch: int):
        self.model.train()
        epoch_losses = {k: 0.0 for k in ['total', 'mask', 'skeleton', 'consistency']}
        epoch_metrics = {k: 0.0 for k in ['mask_dice', 'skeleton_dice']}
        num_batches = 0
        
        for batch_idx, (images, targets) in enumerate(self.train_loader):
            images = images.to(self.device, non_blocking=True)
            targets = {k: v.to(self.device, non_blocking=True) for k, v in targets.items()}
            
            # Forward
            with autocast(enabled=self.cfg.USE_AMP):
                outputs = self.model(images)
                losses = self.criterion(outputs, targets, epoch=epoch)
                loss = losses['total'] / self.cfg.GRADIENT_ACCUMULATION_STEPS
            
            # Backward
            if self.cfg.USE_AMP and self.scaler:
                self.scaler.scale(loss).backward()
            else:
                loss.backward()
            
            # Step
            if (batch_idx + 1) % self.cfg.GRADIENT_ACCUMULATION_STEPS == 0:
                if self.cfg.USE_AMP and self.scaler:
                    self.scaler.step(self.optimizer)
                    self.scaler.update()
                else:
                    self.optimizer.step()
                self.optimizer.zero_grad()
            
            # Accumulate
            for k in epoch_losses:
                if k in losses:
                    epoch_losses[k] += losses[k].item()
            
            with torch.no_grad():
                md = SegmentationMetrics.dice_score(outputs['mask'], targets['mask'])
                sd = SegmentationMetrics.dice_score(outputs['skeleton'], targets['skeleton'])
                epoch_metrics['mask_dice'] += md
                epoch_metrics['skeleton_dice'] += sd
            
            num_batches += 1
            
            if (batch_idx + 1) % 10 == 0:
                print(f"  E{epoch} B{batch_idx+1}/{len(self.train_loader)} | "
                      f"Loss: {losses['total'].item():.3f} | "
                      f"MaskD: {md:.3f} | SkelD: {sd:.3f}")
        
        for k in epoch_losses:
            epoch_losses[k] /= num_batches
        for k in epoch_metrics:
            epoch_metrics[k] /= num_batches
        
        return epoch_losses, epoch_metrics
    
    @torch.no_grad()
    def validate(self, epoch: int):
        self.model.eval()
        val_losses = {k: 0.0 for k in ['total', 'mask', 'skeleton', 'consistency']}
        val_metrics = {k: 0.0 for k in ['mask_dice', 'skeleton_dice', 'mask_iou', 'skeleton_iou']}
        num_batches = 0
        
        for images, targets in self.val_loader:
            images = images.to(self.device, non_blocking=True)
            targets = {k: v.to(self.device, non_blocking=True) for k, v in targets.items()}
            
            with autocast(enabled=self.cfg.USE_AMP):
                outputs = self.model(images)
                losses = self.criterion(outputs, targets, epoch=epoch)
            
            for k in val_losses:
                if k in losses:
                    val_losses[k] += losses[k].item()
            
            md = SegmentationMetrics.dice_score(outputs['mask'], targets['mask'])
            sd = SegmentationMetrics.dice_score(outputs['skeleton'], targets['skeleton'])
            mi = SegmentationMetrics.iou_score(outputs['mask'], targets['mask'])
            si = SegmentationMetrics.iou_score(outputs['skeleton'], targets['skeleton'])
            
            val_metrics['mask_dice'] += md
            val_metrics['skeleton_dice'] += sd
            val_metrics['mask_iou'] += mi
            val_metrics['skeleton_iou'] += si
            
            num_batches += 1
        
        for k in val_losses:
            val_losses[k] /= num_batches
        for k in val_metrics:
            val_metrics[k] /= num_batches
        
        return val_losses, val_metrics
    
    def save_checkpoint(self, epoch: int, is_best: bool = False):
        checkpoint = {
            'epoch': epoch,
            'model_state_dict': self.model.module.state_dict() if hasattr(self.model, 'module') else self.model.state_dict(),
            'optimizer_state_dict': self.optimizer.state_dict(),
            'scheduler_state_dict': self.scheduler.state_dict(),
            'best_val_dice': self.best_val_dice,
            'history': self.history,
        }
        
        path = os.path.join(self.cfg.CHECKPOINT_DIR, f"{self.cfg.EXPERIMENT_NAME}_v2_epoch{epoch}.pt")
        torch.save(checkpoint, path)
        
        if is_best:
            best_path = os.path.join(self.cfg.CHECKPOINT_DIR, f"{self.cfg.EXPERIMENT_NAME}_v2_best.pt")
            torch.save(checkpoint, best_path)
            print(f"  NEW BEST: {best_path}")
    
    def fit(self, epochs: int):
        print(f"\n{'='*60}")
        print("TRAINING V2 — Improved Strategy")
        print(f"{'='*60}")
        
        for epoch in range(1, epochs + 1):
            start = time.time()
            
            # Update LR
            current_lr = self.scheduler.step(epoch)
            
            # Train
            train_losses, train_metrics = self.train_epoch(epoch)
            
            # Validate
            val_losses, val_metrics = self.validate(epoch)
            
            # Log
            print(f"\n{'='*60}")
            print(f"EPOCH {epoch}/{epochs} | LR: {current_lr:.2e} | Time: {time.time()-start:.1f}s")
            print(f"{'='*60}")
            print(f"Train — Loss: {train_losses['total']:.4f} | "
                  f"MaskD: {train_metrics['mask_dice']:.4f} | "
                  f"SkelD: {train_metrics['skeleton_dice']:.4f}")
            print(f"Val   — Loss: {val_losses['total']:.4f} | "
                  f"MaskD: {val_metrics['mask_dice']:.4f} | "
                  f"SkelD: {val_metrics['skeleton_dice']:.4f} | "
                  f"MaskIoU: {val_metrics['mask_iou']:.4f}")
            
            # History
            self.history['epoch'].append(epoch)
            self.history['train_loss'].append(train_losses['total'])
            self.history['val_loss'].append(val_losses['total'])
            self.history['val_mask_dice'].append(val_metrics['mask_dice'])
            self.history['val_skeleton_dice'].append(val_metrics['skeleton_dice'])
            self.history['lr'].append(current_lr)
            
            # Early stopping
            current_dice = val_metrics['mask_dice']
            if current_dice > self.best_val_dice:
                self.best_val_dice = current_dice
                self.epochs_no_improve = 0
                self.save_checkpoint(epoch, is_best=True)
            else:
                self.epochs_no_improve += 1
            
            if self.epochs_no_improve >= self.cfg.EARLY_STOPPING_PATIENCE:
                print(f"\nEarly stopping at epoch {epoch}")
                break
            
            if epoch % 10 == 0:
                self.save_checkpoint(epoch)
        
        self.save_checkpoint(epochs)
        
        # Save history
        with open(os.path.join(self.cfg.LOG_DIR, 'training_history_v2.json'), 'w') as f:
            json.dump(self.history, f, indent=2)
        
        return self.history

# ============================================================
# 9.5 BUILD & TRAIN
# ============================================================

print("=" * 60)
print("BUILDING IMPROVED TRAINING PIPELINE")
print("=" * 60)

# New model (or reload from checkpoint)
model_v2 = TopoEyeNet(
    encoder_name="resnet50",
    encoder_weights="imagenet",
    in_channels=3,
    mask_head=True,
    skeleton_head=True,
).to(device)

if torch.cuda.device_count() > 1:
    model_v2 = nn.DataParallel(model_v2, device_ids=[0, 1])

# Optimizer with higher LR
optimizer_v2 = torch.optim.AdamW(
    model_v2.parameters(),
    lr=1e-3,  # Higher peak LR
    weight_decay=cfg.WEIGHT_DECAY,
)

# Scheduler with warmup
scheduler_v2 = CosineWarmupScheduler(
    optimizer=optimizer_v2,
    warmup_epochs=10,
    max_epochs=100,
    base_lr=1e-5,      # Start very low
    max_lr=1e-3,       # Peak LR
    min_lr=1e-6,
)

# Improved criterion
criterion_v2 = TopoEyeLossV2(
    lambda_mask=1.0,
    lambda_skeleton=1.0,  # Equal weight now
    lambda_consistency=0.3,
    consistency_warmup=10,
)

scaler_v2 = GradScaler() if cfg.USE_AMP else None

print(f"Peak LR: 1e-3 | Warmup: 10 epochs | Consistency warmup: 10 epochs")

# Train
trainer_v2 = TrainerV2(
    model=model_v2,
    criterion=criterion_v2,
    optimizer=optimizer_v2,
    scheduler=scheduler_v2,
    train_loader=rm_train_loader,
    val_loader=rm_val_loader,
    device=device,
    cfg=cfg,
    scaler=scaler_v2,
)

# Run 50 epochs (adjust as needed)
EPOCHS_V2 = 50
history_v2 = trainer_v2.fit(epochs=EPOCHS_V2)

# Plot
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0, 0].plot(history_v2['epoch'], history_v2['train_loss'], label='Train', marker='o')
axes[0, 0].plot(history_v2['epoch'], history_v2['val_loss'], label='Val', marker='o')
axes[0, 0].set_title('Loss')
axes[0, 0].legend(); axes[0, 0].grid(True)

axes[0, 1].plot(history_v2['epoch'], history_v2['val_mask_dice'], label='Mask Dice', marker='o', color='blue')
axes[0, 1].plot(history_v2['epoch'], history_v2['val_skeleton_dice'], label='Skeleton Dice', marker='o', color='red')
axes[0, 1].set_title('Validation Dice')
axes[0, 1].legend(); axes[0, 1].grid(True)

axes[1, 0].plot(history_v2['epoch'], history_v2['lr'], marker='o', color='green')
axes[1, 0].set_title('Learning Rate')
axes[1, 0].set_yscale('log'); axes[1, 0].grid(True)

# Consistency weight over epochs
cons_weights = [min(1.0, e/10) * 0.3 for e in history_v2['epoch']]
axes[1, 1].plot(history_v2['epoch'], cons_weights, marker='o', color='orange')
axes[1, 1].set_title('Consistency Loss Weight (Curriculum)')
axes[1, 1].grid(True)

plt.suptitle("TopoEyeNet V2 Training — Improved Strategy", fontsize=14, fontweight='bold')
plt.tight_layout()
save_path = os.path.join(cfg.FIGURES_DIR, "training_curves_v2.png")
plt.savefig(save_path, dpi=150, bbox_inches='tight')
print(f"\nSaved: {save_path}")
plt.show()

get_gpu_memory()
print(f"\n{'='*60}")
print("CELL 9 COMPLETE")
print(f"{'='*60}")
```

```python
# ============================================================
# CELL 10: EYEPACS PRETRAINING PIPELINE (FIXED)
# ============================================================
# Fixed: DataParallel .get_encoder() access, storage cleanup,
# minimal checkpointing to stay within 20GB budget.

import os
import numpy as np
from PIL import Image
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.cuda.amp import autocast, GradScaler
import glob
import random
import time

# ============================================================
# 10.0 STORAGE CLEANUP (run first)
# ============================================================

def cleanup_storage():
    """Remove intermediate files to free Kaggle storage."""
    print("Cleaning up storage...")
    
    # Remove old epoch checkpoints (keep only best/final)
    checkpoint_dir = cfg.CHECKPOINT_DIR
    if os.path.exists(checkpoint_dir):
        removed = 0
        for f in glob.glob(os.path.join(checkpoint_dir, "*_epoch[0-9]*.pt")):
            if "best" not in f and "final" not in f and "pretrained" not in f:
                os.remove(f)
                removed += 1
        if removed:
            print(f"  Removed {removed} old checkpoints")
    
    # Remove old figures
    figures_dir = cfg.FIGURES_DIR
    if os.path.exists(figures_dir):
        for f in glob.glob(os.path.join(figures_dir, "training_curves_*.png")):
            if "v2" not in f:  # keep latest
                os.remove(f)
    
    # Check usage
    total = 0
    for dirpath, dirnames, filenames in os.walk("/kaggle/working"):
        for f in filenames:
            fp = os.path.join(dirpath, f)
            if os.path.exists(fp):
                total += os.path.getsize(fp)
    
    used_gb = total / 1e9
    print(f"  Storage used: {used_gb:.2f} GB / 20 GB ({used_gb/20*100:.1f}%)")
    return used_gb

cleanup_storage()

# ============================================================
# 10.1 MASKED AUTOENCODER
# ============================================================

class MaskedAutoencoder(nn.Module):
    def __init__(self, encoder_name: str = "resnet50", mask_ratio: float = 0.75):
        super().__init__()
        self.mask_ratio = mask_ratio
        
        self.encoder = smp.encoders.get_encoder(
            name=encoder_name,
            in_channels=3,
            depth=5,
            weights="imagenet",
        )
        
        encoder_channels = self.encoder.out_channels[-1]
        
        self.decoder = nn.Sequential(
            nn.Conv2d(encoder_channels, 512, 1),
            nn.BatchNorm2d(512),
            nn.ReLU(inplace=True),
            nn.ConvTranspose2d(512, 256, 4, 2, 1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.ConvTranspose2d(256, 128, 4, 2, 1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.ConvTranspose2d(128, 64, 4, 2, 1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.ConvTranspose2d(64, 32, 4, 2, 1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.ConvTranspose2d(32, 3, 4, 2, 1),
        )
        
        self._init_weights()
    
    def _init_weights(self):
        for m in self.decoder.modules():
            if isinstance(m, (nn.Conv2d, nn.ConvTranspose2d)):
                nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)
    
    def random_mask(self, x: torch.Tensor) -> tuple:
        B, C, H, W = x.shape
        patch_size = 16
        num_patches_h = H // patch_size
        num_patches_w = W // patch_size
        total_patches = num_patches_h * num_patches_w
        num_masked = int(total_patches * self.mask_ratio)
        
        mask = torch.ones(B, 1, num_patches_h, num_patches_w, device=x.device)
        for b in range(B):
            flat_indices = torch.randperm(total_patches)[:num_masked]
            h_indices = flat_indices // num_patches_w
            w_indices = flat_indices % num_patches_w
            mask[b, 0, h_indices, w_indices] = 0
        
        mask_upsampled = F.interpolate(mask, size=(H, W), mode='nearest')
        masked_x = x * mask_upsampled
        return masked_x, mask_upsampled
    
    def forward(self, x: torch.Tensor) -> tuple:
        masked_x, mask = self.random_mask(x)
        features = self.encoder(masked_x)
        deepest = features[-1]
        recon = self.decoder(deepest)
        return recon, mask
    
    def get_encoder(self):
        return self.encoder

# ============================================================
# 10.2 EYEPACS DATASET
# ============================================================

class EyePACSDataset(Dataset):
    def __init__(self, image_dir: str, img_size: int = 512, max_samples: int = None):
        self.image_paths = []
        for root, _, files in os.walk(image_dir):
            for f in files:
                if f.lower().endswith(('.jpg', '.jpeg', '.png', '.tif')):
                    self.image_paths.append(os.path.join(root, f))
        
        if max_samples:
            self.image_paths = self.image_paths[:max_samples]
        
        self.img_size = img_size
        self.num_samples = len(self.image_paths)
        print(f"EyePACS dataset: {self.num_samples} images")
    
    def __len__(self):
        return self.num_samples
    
    def __getitem__(self, idx):
        path = self.image_paths[idx]
        img = Image.open(path).convert('RGB')
        img = img.resize((self.img_size, self.img_size), Image.BILINEAR)
        img = np.array(img).astype(np.float32) / 255.0
        
        mean = np.array([0.485, 0.456, 0.406])
        std = np.array([0.229, 0.224, 0.225])
        img = (img - mean) / std
        img = torch.from_numpy(img).permute(2, 0, 1).float()
        return img

# ============================================================
# 10.3 PRETRAINER (FIXED for DataParallel)
# ============================================================

class Pretrainer:
    def __init__(self, model, optimizer, scheduler, train_loader, device, cfg):
        self.model = model
        self.optimizer = optimizer
        self.scheduler = scheduler
        self.train_loader = train_loader
        self.device = device
        self.cfg = cfg
        self.history = {'epoch': [], 'loss': [], 'mse': [], 'psnr': []}
    
    def train_epoch(self, epoch: int):
        self.model.train()
        total_loss = 0.0
        total_mse = 0.0
        num_batches = 0
        
        for batch_idx, images in enumerate(self.train_loader):
            images = images.to(self.device, non_blocking=True)
            
            with autocast(enabled=self.cfg.USE_AMP):
                recon, mask = self.model(images)
                loss_visible = F.mse_loss(recon * mask, images * mask)
                loss_masked = F.mse_loss(recon * (1 - mask), images * (1 - mask))
                loss = loss_visible + 2.0 * loss_masked
                
                tv_h = torch.abs(recon[:, :, 1:, :] - recon[:, :, :-1, :]).mean()
                tv_w = torch.abs(recon[:, :, :, 1:] - recon[:, :, :, :-1]).mean()
                loss += 0.01 * (tv_h + tv_w)
            
            if self.cfg.USE_AMP:
                scaler.scale(loss).backward()
                scaler.step(self.optimizer)
                scaler.update()
            else:
                loss.backward()
                self.optimizer.step()
            
            self.optimizer.zero_grad()
            
            total_loss += loss.item()
            with torch.no_grad():
                mse = F.mse_loss(recon, images).item()
                total_mse += mse
            
            num_batches += 1
            
            if (batch_idx + 1) % 100 == 0:
                psnr = -10 * np.log10(mse + 1e-10)
                print(f"  E{epoch} B{batch_idx+1}/{len(self.train_loader)} | "
                      f"Loss: {loss.item():.4f} | PSNR: {psnr:.2f}")
        
        avg_loss = total_loss / num_batches
        avg_mse = total_mse / num_batches
        avg_psnr = -10 * np.log10(avg_mse + 1e-10)
        return avg_loss, avg_mse, avg_psnr
    
    def fit(self, epochs: int):
        print(f"\n{'='*60}")
        print("EYEPACS PRETRAINING")
        print(f"{'='*60}")
        print(f"Epochs: {epochs} | Batches: {len(self.train_loader)}")
        
        for epoch in range(1, epochs + 1):
            start = time.time()
            loss, mse, psnr = self.train_epoch(epoch)
            self.scheduler.step()
            
            print(f"\n{'='*60}")
            print(f"EPOCH {epoch}/{epochs} | Time: {time.time()-start:.1f}s")
            print(f"  Loss: {loss:.4f} | MSE: {mse:.4f} | PSNR: {psnr:.2f} dB")
            
            self.history['epoch'].append(epoch)
            self.history['loss'].append(loss)
            self.history['mse'].append(mse)
            self.history['psnr'].append(psnr)
            
            # Only save at end and halfway to save space
            if epoch == epochs or epoch == epochs // 2:
                self.save_checkpoint(epoch, is_final=(epoch == epochs))
            
            # Cleanup if storage getting full
            if epoch % 3 == 0:
                cleanup_storage()
        
        return self.history
    
    def save_checkpoint(self, epoch: int, is_final: bool = False):
        # FIXED: Handle DataParallel
        if hasattr(self.model, 'module'):
            encoder = self.model.module.get_encoder()
        else:
            encoder = self.model.get_encoder()
        
        state = {
            'epoch': epoch,
            'encoder_state_dict': encoder.state_dict(),
            'optimizer_state_dict': self.optimizer.state_dict(),
        }
        
        name = "pretrained_encoder_final.pt" if is_final else f"pretrained_encoder_e{epoch}.pt"
        path = os.path.join(self.cfg.CHECKPOINT_DIR, name)
        torch.save(state, path)
        size_mb = os.path.getsize(path) / 1e6
        print(f"  Saved: {path} ({size_mb:.1f} MB)")

# ============================================================
# 10.4 BUILD & RUN
# ============================================================

print("=" * 60)
print("BUILDING EYEPACS PRETRAINING")
print("=" * 60)

# Use 5000 images for quick pretraining (was working well)
EYEPACS_MAX_SAMPLES = 5000

eyepacs_dataset = EyePACSDataset(
    image_dir=cfg.EYEPACS_IMAGE_DIR,
    img_size=cfg.IMG_SIZE,
    max_samples=EYEPACS_MAX_SAMPLES,
)

eyepacs_loader = DataLoader(
    eyepacs_dataset,
    batch_size=cfg.BATCH_SIZE * 2,  # 8 per batch
    shuffle=True,
    num_workers=cfg.NUM_WORKERS,
    pin_memory=True,
    drop_last=True,
)

print(f"EyePACS loader: {len(eyepacs_loader)} batches")

# Model
mae_model = MaskedAutoencoder(encoder_name="resnet50", mask_ratio=0.75)
mae_model = mae_model.to(device)

if torch.cuda.device_count() > 1:
    mae_model = nn.DataParallel(mae_model, device_ids=[0, 1])

# Optimizer
pretrain_optimizer = torch.optim.AdamW(mae_model.parameters(), lr=1e-4, weight_decay=1e-4)
pretrain_scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(pretrain_optimizer, T_max=10, eta_min=1e-6)

# Run remaining 5 epochs (already did 5, continue from there or restart)
# If you want to continue from epoch 5, load the checkpoint. Otherwise fresh start.
PRETRAIN_EPOCHS = 5  # 5 more epochs = 10 total

pretrainer = Pretrainer(
    model=mae_model,
    optimizer=pretrain_optimizer,
    scheduler=pretrain_scheduler,
    train_loader=eyepacs_loader,
    device=device,
    cfg=cfg,
)

pretrain_history = pretrainer.fit(epochs=PRETRAIN_EPOCHS)

# ============================================================
# 10.5 VISUALIZE
# ============================================================

print(f"\n{'='*60}")
print("VISUALIZING RECONSTRUCTIONS")
print(f"{'='*60}")

mae_model.eval()
with torch.no_grad():
    sample_images = next(iter(eyepacs_loader))[:4].to(device)
    recon, mask = mae_model(sample_images)
    
    mean = torch.tensor([0.485, 0.456, 0.406]).view(1, 3, 1, 1).to(device)
    std = torch.tensor([0.229, 0.224, 0.225]).view(1, 3, 1, 1).to(device)
    
    orig = sample_images * std + mean
    recon = recon * std + mean
    masked = (sample_images * mask) * std + mean
    
    fig, axes = plt.subplots(3, 4, figsize=(16, 12))
    for i in range(4):
        axes[0, i].imshow(orig[i].cpu().permute(1, 2, 0).clamp(0, 1))
        axes[0, i].set_title("Original")
        axes[0, i].axis('off')
        
        axes[1, i].imshow(masked[i].cpu().permute(1, 2, 0).clamp(0, 1))
        axes[1, i].set_title("Masked (75%)")
        axes[1, i].axis('off')
        
        axes[2, i].imshow(recon[i].cpu().permute(1, 2, 0).clamp(0, 1))
        axes[2, i].set_title("Reconstruction")
        axes[2, i].axis('off')
    
    plt.suptitle("EyePACS Pretraining — Reconstructions", fontsize=14, fontweight='bold')
    plt.tight_layout()
    save_path = os.path.join(cfg.FIGURES_DIR, "pretrain_reconstructions.png")
    plt.savefig(save_path, dpi=150, bbox_inches='tight')
    print(f"Saved: {save_path}")
    plt.show()

# Plot curves
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].plot(pretrain_history['epoch'], pretrain_history['loss'], marker='o')
axes[0].set_title('Loss'); axes[0].set_xlabel('Epoch'); axes[0].grid(True)

axes[1].plot(pretrain_history['epoch'], pretrain_history['psnr'], marker='o', color='green')
axes[1].set_title('PSNR (dB)'); axes[1].set_xlabel('Epoch'); axes[1].grid(True)

plt.suptitle("Pretraining Curves", fontsize=14, fontweight='bold')
plt.tight_layout()
save_path = os.path.join(cfg.FIGURES_DIR, "pretrain_curves.png")
plt.savefig(save_path, dpi=150, bbox_inches='tight')
print(f"Saved: {save_path}")
plt.show()

cleanup_storage()
get_gpu_memory()
print(f"\n{'='*60}")
print("CELL 10 COMPLETE")
print(f"{'='*60}")
```

```python
# ============================================================
# CELL 11: FINE-TUNING WITH PRETRAINED ENCODER
# ============================================================
# Load pretrained encoder from EyePACS, add multi-task heads,
# fine-tune on RetinoMix-5 with mask + skeleton supervision.

import torch
import torch.nn as nn
import segmentation_models_pytorch as smp
from torch.cuda.amp import autocast, GradScaler
import time
import json

# ============================================================
# 11.1 BUILD TOPOEYENET WITH PRETRAINED ENCODER
# ============================================================

class TopoEyeNetPretrained(nn.Module):
    """
    TopoEyeNet with pretrained encoder loaded from MAE.
    Same architecture as before, but encoder starts with fundus knowledge.
    """

    def __init__(self,
                 encoder_name: str = "resnet50",
                 in_channels: int = 3,
                 mask_head: bool = True,
                 skeleton_head: bool = True,
                 pretrained_path: str = None):
        super().__init__()

        self.mask_head_enabled = mask_head
        self.skeleton_head_enabled = skeleton_head

        # Encoder (same as MAE)
        self.encoder = smp.encoders.get_encoder(
            name=encoder_name,
            in_channels=in_channels,
            depth=5,
            weights=None,  # Don't load ImageNet, we'll load our pretrained weights
        )

        # Load pretrained encoder weights
        if pretrained_path and os.path.exists(pretrained_path):
            print(f"Loading pretrained encoder from: {pretrained_path}")
            checkpoint = torch.load(pretrained_path, map_location='cpu')
            encoder_state = checkpoint['encoder_state_dict']

            # Handle DataParallel prefix if present
            new_state = {}
            for k, v in encoder_state.items():
                # Remove 'module.' prefix if present
                name = k.replace('module.', '') if k.startswith('module.') else k
                new_state[name] = v

            # NOTE: SMP's EncoderMixin.load_state_dict override pops 'fc.*' keys
            # and calls super().load_state_dict() WITHOUT returning the result,
            # so it always returns None here -- don't try to unpack it.
            own_keys = set(self.encoder.state_dict().keys())
            loaded_keys = set(new_state.keys())
            missing = own_keys - loaded_keys
            unexpected = loaded_keys - own_keys

            self.encoder.load_state_dict(new_state, strict=False)
            if missing:
                print(f"  Missing keys: {len(missing)}")
            if unexpected:
                print(f"  Unexpected keys: {len(unexpected)}")
            print(f"  Pretrained encoder loaded successfully")
        else:
            print(f"WARNING: No pretrained weights found at {pretrained_path}")
            print(f"  Using random initialization")

        encoder_channels = self.encoder.out_channels

        # Decoder (same as before)
        self.decoder = smp.decoders.unet.decoder.UnetDecoder(
            encoder_channels=encoder_channels,
            decoder_channels=(256, 128, 64, 32, 16),
            n_blocks=5,
            use_batchnorm=True,
            attention_type=None,
            center=False,
        )

        # Heads
        if mask_head:
            self.mask_head = nn.Sequential(
                nn.Conv2d(16, 16, 3, padding=1),
                nn.BatchNorm2d(16),
                nn.ReLU(inplace=True),
                nn.Conv2d(16, 1, 1),
            )
        if skeleton_head:
            self.skeleton_head = nn.Sequential(
                nn.Conv2d(16, 16, 3, padding=1),
                nn.BatchNorm2d(16),
                nn.ReLU(inplace=True),
                nn.Conv2d(16, 1, 1),
            )

        self._init_decoder_heads()

    def _init_decoder_heads(self):
        # Only initialize decoder and heads (encoder is pretrained)
        for m in self.decoder.modules():
            if isinstance(m, (nn.Conv2d, nn.ConvTranspose2d)):
                nn.init.kaiming_normal_(m.weight, mode='fan_out', nonlinearity='relu')
                if m.bias is not None:
                    nn.init.constant_(m.bias, 0)
            elif isinstance(m, nn.BatchNorm2d):
                nn.init.constant_(m.weight, 1)
                nn.init.constant_(m.bias, 0)

        for m in [self.mask_head, self.skeleton_head]:
            if m is not None:
                for layer in m.modules():
                    if isinstance(layer, nn.Conv2d):
                        nn.init.kaiming_normal_(layer.weight, mode='fan_out', nonlinearity='relu')
                        if layer.bias is not None:
                            nn.init.constant_(layer.bias, 0)

    def forward(self, x):
        features = self.encoder(x)
        decoder_output = self.decoder(*features)

        outputs = {}
        if self.mask_head_enabled:
            outputs['mask'] = self.mask_head(decoder_output)
        if self.skeleton_head_enabled:
            outputs['skeleton'] = self.skeleton_head(decoder_output)

        return outputs

# ============================================================
# 11.2 LOSS (same improved version from Cell 9)
# ============================================================

class TopoEyeLossV2(nn.Module):
    def __init__(self,
                 lambda_mask: float = 1.0,
                 lambda_skeleton: float = 1.0,
                 lambda_consistency: float = 0.3,
                 consistency_warmup: int = 10):
        super().__init__()
        self.lambda_mask = lambda_mask
        self.lambda_skeleton = lambda_skeleton
        self.lambda_consistency_full = lambda_consistency
        self.consistency_warmup = consistency_warmup

        self.mask_loss = TverskyLoss(alpha=0.3, beta=0.7)
        self.skeleton_loss = nn.ModuleDict({
            'focal': FocalLoss(alpha=0.25, gamma=2.0),
            'dice': WeightedDiceLoss(pos_weight=50.0),
        })
        self.consistency = MaskSkeletonAgreementLoss()

    def get_consistency_weight(self, epoch: int) -> float:
        if epoch <= 0:
            return 0.0
        if epoch >= self.consistency_warmup:
            return self.lambda_consistency_full
        return self.lambda_consistency_full * (epoch / self.consistency_warmup)

    def forward(self, predictions: dict, targets: dict, epoch: int = 0) -> dict:
        losses = {}
        total_loss = 0.0

        if 'mask' in predictions and 'mask' in targets:
            losses['mask'] = self.mask_loss(predictions['mask'], targets['mask'])
            total_loss += self.lambda_mask * losses['mask']

        if 'skeleton' in predictions and 'skeleton' in targets:
            focal = self.skeleton_loss['focal'](predictions['skeleton'], targets['skeleton'])
            dice = self.skeleton_loss['dice'](predictions['skeleton'], targets['skeleton'])
            losses['skeleton_focal'] = focal
            losses['skeleton_dice'] = dice
            losses['skeleton'] = 0.5 * focal + 0.5 * dice
            total_loss += self.lambda_skeleton * losses['skeleton']

        if 'mask' in predictions and 'skeleton' in predictions:
            cons_weight = self.get_consistency_weight(epoch)
            losses['consistency'] = cons_weight * self.consistency(predictions['mask'], predictions['skeleton'])
            total_loss += losses['consistency']

        losses['total'] = total_loss
        return losses

# ============================================================
# 11.3 TRAINER (same as Cell 9 but with pretrained model)
# ============================================================

class TrainerV3:
    def __init__(self, model, criterion, optimizer, scheduler,
                 train_loader, val_loader, device, cfg, scaler=None):
        self.model = model
        self.criterion = criterion
        self.optimizer = optimizer
        self.scheduler = scheduler
        self.train_loader = train_loader
        self.val_loader = val_loader
        self.device = device
        self.cfg = cfg
        self.scaler = scaler

        self.best_val_dice = 0.0
        self.epochs_no_improve = 0
        self.history = {k: [] for k in ['epoch', 'train_loss', 'val_loss',
                                         'val_mask_dice', 'val_skeleton_dice', 'lr']}

        os.makedirs(cfg.CHECKPOINT_DIR, exist_ok=True)

    def train_epoch(self, epoch: int):
        self.model.train()
        epoch_losses = {k: 0.0 for k in ['total', 'mask', 'skeleton', 'consistency']}
        epoch_metrics = {k: 0.0 for k in ['mask_dice', 'skeleton_dice']}
        num_batches = 0

        for batch_idx, (images, targets) in enumerate(self.train_loader):
            images = images.to(self.device, non_blocking=True)
            targets = {k: v.to(self.device, non_blocking=True) for k, v in targets.items()}

            with autocast(enabled=self.cfg.USE_AMP):
                outputs = self.model(images)
                losses = self.criterion(outputs, targets, epoch=epoch)
                loss = losses['total'] / self.cfg.GRADIENT_ACCUMULATION_STEPS

            if self.cfg.USE_AMP and self.scaler:
                self.scaler.scale(loss).backward()
            else:
                loss.backward()

            if (batch_idx + 1) % self.cfg.GRADIENT_ACCUMULATION_STEPS == 0:
                if self.cfg.USE_AMP and self.scaler:
                    self.scaler.step(self.optimizer)
                    self.scaler.update()
                else:
                    self.optimizer.step()
                self.optimizer.zero_grad()

            for k in epoch_losses:
                if k in losses:
                    epoch_losses[k] += losses[k].item()

            with torch.no_grad():
                md = SegmentationMetrics.dice_score(outputs['mask'], targets['mask'])
                sd = SegmentationMetrics.dice_score(outputs['skeleton'], targets['skeleton'])
                epoch_metrics['mask_dice'] += md
                epoch_metrics['skeleton_dice'] += sd

            num_batches += 1

            if (batch_idx + 1) % self.cfg.CLEAR_CACHE_EVERY_N_BATCHES == 0:
                clear_gpu_cache()

            if (batch_idx + 1) % 10 == 0:
                print(f"  E{epoch} B{batch_idx+1}/{len(self.train_loader)} | "
                      f"Loss: {losses['total'].item():.3f} | "
                      f"MaskD: {md:.3f} | SkelD: {sd:.3f}")

        for k in epoch_losses:
            epoch_losses[k] /= num_batches
        for k in epoch_metrics:
            epoch_metrics[k] /= num_batches

        return epoch_losses, epoch_metrics

    @torch.no_grad()
    def validate(self, epoch: int):
        self.model.eval()
        val_losses = {k: 0.0 for k in ['total', 'mask', 'skeleton', 'consistency']}
        val_metrics = {k: 0.0 for k in ['mask_dice', 'skeleton_dice', 'mask_iou', 'skeleton_iou']}
        num_batches = 0

        for images, targets in self.val_loader:
            images = images.to(self.device, non_blocking=True)
            targets = {k: v.to(self.device, non_blocking=True) for k, v in targets.items()}

            with autocast(enabled=self.cfg.USE_AMP):
                outputs = self.model(images)
                losses = self.criterion(outputs, targets, epoch=epoch)

            for k in val_losses:
                if k in losses:
                    val_losses[k] += losses[k].item()

            md = SegmentationMetrics.dice_score(outputs['mask'], targets['mask'])
            sd = SegmentationMetrics.dice_score(outputs['skeleton'], targets['skeleton'])
            mi = SegmentationMetrics.iou_score(outputs['mask'], targets['mask'])
            si = SegmentationMetrics.iou_score(outputs['skeleton'], targets['skeleton'])

            val_metrics['mask_dice'] += md
            val_metrics['skeleton_dice'] += sd
            val_metrics['mask_iou'] += mi
            val_metrics['skeleton_iou'] += si

            num_batches += 1

        for k in val_losses:
            val_losses[k] /= num_batches
        for k in val_metrics:
            val_metrics[k] /= num_batches

        return val_losses, val_metrics

    def save_checkpoint(self, epoch: int, is_best: bool = False):
        checkpoint = {
            'epoch': epoch,
            'model_state_dict': self.model.module.state_dict() if hasattr(self.model, 'module') else self.model.state_dict(),
            'optimizer_state_dict': self.optimizer.state_dict(),
            'scheduler_state_dict': self.scheduler.state_dict(),
            'best_val_dice': self.best_val_dice,
            'history': self.history,
        }

        path = os.path.join(self.cfg.CHECKPOINT_DIR, f"topoeyenet_pretrained_epoch{epoch}.pt")
        torch.save(checkpoint, path)
        print(f"  Saved checkpoint: {path}")

        if is_best:
            best_path = os.path.join(self.cfg.CHECKPOINT_DIR, "topoeyenet_pretrained_best.pt")
            torch.save(checkpoint, best_path)
            print(f"  Saved BEST checkpoint: {best_path}")

    def fit(self, epochs: int):
        print(f"\n{'='*60}")
        print("STARTING FINE-TUNING (Pretrained Encoder)")
        print(f"{'='*60}")
        print(f"Epochs: {epochs}")
        print(f"Train batches: {len(self.train_loader)}")
        print(f"Val batches: {len(self.val_loader)}")
        print(f"AMP: {self.cfg.USE_AMP}")
        print(f"Gradient Accumulation: {self.cfg.GRADIENT_ACCUMULATION_STEPS}")

        final_epoch = 0
        for epoch in range(1, epochs + 1):
            final_epoch = epoch
            start_time = time.time()

            train_losses, train_metrics = self.train_epoch(epoch)

            if epoch % self.cfg.VAL_FREQ == 0:
                val_losses, val_metrics = self.validate(epoch)
                self.scheduler.step()

                print(f"\n{'='*60}")
                print(f"EPOCH {epoch}/{epochs} | LR: {self.optimizer.param_groups[0]['lr']:.2e} | "
                      f"Time: {time.time()-start_time:.1f}s")
                print(f"{'='*60}")
                print(f"Train — Loss: {train_losses['total']:.4f} | "
                      f"MaskD: {train_metrics['mask_dice']:.4f} | SkelD: {train_metrics['skeleton_dice']:.4f}")
                print(f"Val   — Loss: {val_losses['total']:.4f} | "
                      f"MaskD: {val_metrics['mask_dice']:.4f} | SkelD: {val_metrics['skeleton_dice']:.4f} | "
                      f"MaskIoU: {val_metrics['mask_iou']:.4f}")

                self.history['epoch'].append(epoch)
                self.history['train_loss'].append(train_losses['total'])
                self.history['val_loss'].append(val_losses['total'])
                self.history['val_mask_dice'].append(val_metrics['mask_dice'])
                self.history['val_skeleton_dice'].append(val_metrics['skeleton_dice'])
                self.history['lr'].append(self.optimizer.param_groups[0]['lr'])

                current_dice = val_metrics['mask_dice']
                if current_dice > self.best_val_dice:
                    self.best_val_dice = current_dice
                    self.epochs_no_improve = 0
                    if self.cfg.SAVE_BEST_ONLY:
                        self.save_checkpoint(epoch, is_best=True)
                else:
                    self.epochs_no_improve += 1

                if self.epochs_no_improve >= self.cfg.EARLY_STOPPING_PATIENCE:
                    print(f"\nEarly stopping triggered after {epoch} epochs")
                    break

                if not self.cfg.SAVE_BEST_ONLY:
                    self.save_checkpoint(epoch)
            else:
                self.scheduler.step()

        self.save_checkpoint(final_epoch)

        history_path = os.path.join(self.cfg.LOG_DIR, 'training_history_pretrained.json')
        with open(history_path, 'w') as f:
            json.dump(self.history, f, indent=2)
        print(f"\nSaved history: {history_path}")

        return self.history

 #============================================================
# 11.4 (UPDATED) BUILD MODEL WITH PRETRAINED ENCODER
# ============================================================
 
USE_SINGLE_GPU = True  # <-- toggle this to False to reproduce the old DataParallel setup
 
print("=" * 60)
print("BUILDING TOPOEYENET (PRETRAINED ENCODER) — FULL RUN")
print("=" * 60)
print(f"Single-GPU mode: {USE_SINGLE_GPU}")
 
PRETRAINED_ENCODER_PATH = os.path.join(cfg.CHECKPOINT_DIR, "pretrained_encoder_final.pt")
 
model_pretrained_full = TopoEyeNetPretrained(
    encoder_name="resnet50",
    in_channels=cfg.IN_CHANNELS,
    mask_head=cfg.USE_MASK_HEAD,
    skeleton_head=cfg.USE_SKELETON_HEAD,
    pretrained_path=PRETRAINED_ENCODER_PATH,
)
 
total_params = sum(p.numel() for p in model_pretrained_full.parameters())
print(f"Total parameters: {total_params:,}")
 
if USE_SINGLE_GPU:
    model_pretrained_full = model_pretrained_full.to(torch.device("cuda:0"))
    run_device = torch.device("cuda:0")
    print(f"Model on: {run_device} (single GPU, full batch size {cfg.BATCH_SIZE} for BatchNorm)")
else:
    model_pretrained_full = wrap_dual_gpu(model_pretrained_full)
    run_device = device
    print(f"Model on: {run_device} (DataParallel, effective per-replica batch size "
          f"{cfg.BATCH_SIZE // 2})")
 
# ============================================================
# 11.5 LOSS, OPTIMIZER, SCHEDULER
# ============================================================
 
criterion_full = TopoEyeLossV2(
    lambda_mask=cfg.LAMBDA_MASK,
    lambda_skeleton=cfg.LAMBDA_SKELETON,
    lambda_consistency=cfg.LAMBDA_CONSISTENCY,
    consistency_warmup=10,
)
 
optimizer_full = torch.optim.AdamW(
    model_pretrained_full.parameters(),
    lr=cfg.LEARNING_RATE,
    weight_decay=cfg.WEIGHT_DECAY,
)
 
FT_EPOCHS = cfg.EPOCHS  # full run: 100 epochs
 
scheduler_full = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer_full, T_max=FT_EPOCHS, eta_min=cfg.ETA_MIN
)
 
scaler_full = GradScaler() if cfg.USE_AMP else None
 
print(f"Optimizer: AdamW | LR: {cfg.LEARNING_RATE}")
print(f"Scheduler: Cosine | T_max: {FT_EPOCHS}")
print(f"AMP: {cfg.USE_AMP}")
print(f"Early stopping patience: {cfg.EARLY_STOPPING_PATIENCE}")
 
# ============================================================
# 11.6 SANITY CHECK
# ============================================================
 
print(f"\n{'='*60}")
print("SANITY CHECK: 1 batch")
print(f"{'='*60}")
 
model_pretrained_full.train()
for images, targets in rm_train_loader:
    images = images.to(run_device)
    targets = {k: v.to(run_device) for k, v in targets.items()}
 
    with autocast(enabled=cfg.USE_AMP):
        outputs = model_pretrained_full(images)
        losses = criterion_full(outputs, targets, epoch=0)
 
    print("Forward pass OK")
    print(f"  Loss components: { {k: f'{v.item():.4f}' for k, v in losses.items()} }")
 
    if cfg.USE_AMP and scaler_full:
        scaler_full.scale(losses['total']).backward()
        scaler_full.step(optimizer_full)
        scaler_full.update()
    else:
        losses['total'].backward()
        optimizer_full.step()
 
    optimizer_full.zero_grad()
    print("  Backward pass OK")
    break
 
clear_gpu_cache()
 
# ============================================================
# 11.7 FULL FINE-TUNE (100 epochs)
# ============================================================
 
trainer_full = TrainerV3(
    model=model_pretrained_full,
    criterion=criterion_full,
    optimizer=optimizer_full,
    scheduler=scheduler_full,
    train_loader=rm_train_loader,
    val_loader=rm_val_loader,
    device=run_device,
    cfg=cfg,
    scaler=scaler_full,
)
 
print(f"\n{'='*60}")
print(f"FULL FINE-TUNING — {FT_EPOCHS} EPOCHS — SINGLE_GPU={USE_SINGLE_GPU}")
print(f"{'='*60}")
 
history_full = trainer_full.fit(epochs=FT_EPOCHS)
 
# ============================================================
# 11.8 PLOT & SUMMARY
# ============================================================
 
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
 
axes[0].plot(history_full['epoch'], history_full['train_loss'], label='Train Loss', marker='o', markersize=3)
axes[0].plot(history_full['epoch'], history_full['val_loss'], label='Val Loss', marker='o', markersize=3)
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].set_title(f'Fine-Tuning Loss (single_gpu={USE_SINGLE_GPU})')
axes[0].legend()
axes[0].grid(True)
 
axes[1].plot(history_full['epoch'], history_full['val_mask_dice'], label='Mask Dice', marker='o', markersize=3)
axes[1].plot(history_full['epoch'], history_full['val_skeleton_dice'], label='Skeleton Dice', marker='o', markersize=3)
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Dice Score')
axes[1].set_title(f'Validation Dice (single_gpu={USE_SINGLE_GPU})')
axes[1].legend()
axes[1].grid(True)
 
plt.suptitle(f"TopoEyeNet Full Fine-Tune — {FT_EPOCHS} epochs, single_gpu={USE_SINGLE_GPU}",
             fontsize=14, fontweight='bold')
plt.tight_layout()
suffix = "single_gpu" if USE_SINGLE_GPU else "dataparallel"
save_path = os.path.join(cfg.FIGURES_DIR, f"training_curves_full_{suffix}.png")
plt.savefig(save_path, dpi=150, bbox_inches='tight')
print(f"\nSaved: {save_path}")
plt.show()
 
print(f"\n{'='*60}")
print("FULL FINE-TUNING SUMMARY")
print(f"{'='*60}")
print(f"  Config: {FT_EPOCHS} epochs | single_gpu={USE_SINGLE_GPU}")
print(f"  Best mask Dice:     {trainer_full.best_val_dice:.4f}")
print(f"  Final skeleton Dice: {history_full['val_skeleton_dice'][-1]:.4f}" if history_full['val_skeleton_dice'] else "  N/A")
 
get_gpu_memory()
print(f"\n{'='*60}")
print("CELL 11 (FULL RUN) COMPLETE")
print(f"{'='*60}")
```

```python
# ============================================================
# CELL 12: DIAGNOSTIC BASELINE — VANILLA U-NET (MASK ONLY)
# ============================================================
# Purpose: isolate whether the Dice ceiling (~0.17-0.27) seen in
# Cells 8/9/11 is caused by:
#   (a) the data/pipeline itself (resolution, mask quality, augmentation), or
#   (b) something specific to TopoEyeNet (multi-task heads, consistency
#       loss, DataParallel batch-size-2 BatchNorm instability, etc.)
#
# This cell trains the most standard, well-understood vessel-segmentation
# setup possible: off-the-shelf smp.Unet, single task (mask only),
# plain DiceBCE loss, single GPU (so BatchNorm sees the FULL batch size,
# not a batch of 2 split across DataParallel).
#
# If THIS gets a Dice ceiling similar to TopoEyeNet -> the bug is upstream
# (data/resolution/masks), not in the custom architecture.
# If THIS clears 0.6-0.7+ Dice quickly -> the bug is TopoEyeNet-specific
# (multi-task setup, DataParallel batch norm, or the consistency loss).

import torch
import torch.nn as nn
import segmentation_models_pytorch as smp
from torch.cuda.amp import autocast, GradScaler
import time
import json

# ============================================================
# 12.1 SINGLE-GPU DEVICE (deliberately NOT wrap_dual_gpu)
# ============================================================

baseline_device = torch.device("cuda:0")
print("=" * 60)
print("BASELINE DIAGNOSTIC — VANILLA U-NET (MASK ONLY)")
print("=" * 60)
print(f"Device: {baseline_device} (single GPU, no DataParallel)")
print(f"Batch size: {cfg.BATCH_SIZE} (full batch on one GPU, not split)")

# ============================================================
# 12.2 BUILD PLAIN U-NET
# ============================================================
# resnet34 is a deliberate choice: it's the most commonly reported
# encoder in vessel-segmentation literature (DRIVE/CHASE-DB/STARE
# baselines), so if this can't clear a reasonable Dice on your data,
# that's strong evidence the issue is upstream of any architecture choice.

baseline_model = smp.Unet(
    encoder_name="resnet34",
    encoder_weights="imagenet",
    in_channels=3,
    classes=1,
).to(baseline_device)

total_params = sum(p.numel() for p in baseline_model.parameters())
print(f"Total parameters: {total_params:,}")

# ============================================================
# 12.3 LOSS, OPTIMIZER
# ============================================================
# Reuses DiceBCELoss already defined in Cell 6 -- deliberately the
# simplest loss in the notebook, no Tversky/Focal/consistency terms.

baseline_criterion = DiceBCELoss(dice_weight=0.7, bce_weight=0.3)

baseline_optimizer = torch.optim.AdamW(
    baseline_model.parameters(),
    lr=cfg.LEARNING_RATE,
    weight_decay=cfg.WEIGHT_DECAY,
)

BASELINE_EPOCHS = 20  # short diagnostic run, not a full training budget

baseline_scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    baseline_optimizer, T_max=BASELINE_EPOCHS, eta_min=cfg.ETA_MIN
)

baseline_scaler = GradScaler() if cfg.USE_AMP else None

# ============================================================
# 12.4 SANITY CHECK: 1 BATCH
# ============================================================

print(f"\n{'='*60}")
print("SANITY CHECK: 1 batch")
print(f"{'='*60}")

baseline_model.train()
for images, targets in rm_train_loader:
    images = images.to(baseline_device)
    mask_target = targets['mask'].to(baseline_device)

    with autocast(enabled=cfg.USE_AMP):
        pred = baseline_model(images)
        loss = baseline_criterion(pred, mask_target)

    print("Forward pass OK")
    print(f"  Loss: {loss.item():.4f}")

    if cfg.USE_AMP and baseline_scaler:
        baseline_scaler.scale(loss).backward()
        baseline_scaler.step(baseline_optimizer)
        baseline_scaler.update()
    else:
        loss.backward()
        baseline_optimizer.step()

    baseline_optimizer.zero_grad()
    print("  Backward pass OK")
    break

clear_gpu_cache()

# ============================================================
# 12.5 TRAIN LOOP (mask only, simplified)
# ============================================================

def train_baseline_epoch(model, loader, criterion, optimizer, scaler, cfg, device):
    model.train()
    total_loss, total_dice = 0.0, 0.0
    num_batches = 0

    for batch_idx, (images, targets) in enumerate(loader):
        images = images.to(device, non_blocking=True)
        mask_target = targets['mask'].to(device, non_blocking=True)

        with autocast(enabled=cfg.USE_AMP):
            pred = model(images)
            loss = criterion(pred, mask_target)

        if cfg.USE_AMP and scaler:
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()
        else:
            loss.backward()
            optimizer.step()
        optimizer.zero_grad()

        with torch.no_grad():
            dice = SegmentationMetrics.dice_score(pred, mask_target)

        total_loss += loss.item()
        total_dice += dice
        num_batches += 1

        if (batch_idx + 1) % 10 == 0:
            print(f"  B{batch_idx+1}/{len(loader)} | Loss: {loss.item():.3f} | Dice: {dice:.3f}")

    return total_loss / num_batches, total_dice / num_batches


@torch.no_grad()
def validate_baseline(model, loader, criterion, cfg, device):
    model.eval()
    total_loss, total_dice, total_iou = 0.0, 0.0, 0.0
    num_batches = 0

    for images, targets in loader:
        images = images.to(device, non_blocking=True)
        mask_target = targets['mask'].to(device, non_blocking=True)

        with autocast(enabled=cfg.USE_AMP):
            pred = model(images)
            loss = criterion(pred, mask_target)

        dice = SegmentationMetrics.dice_score(pred, mask_target)
        iou = SegmentationMetrics.iou_score(pred, mask_target)

        total_loss += loss.item()
        total_dice += dice
        total_iou += iou
        num_batches += 1

    return total_loss / num_batches, total_dice / num_batches, total_iou / num_batches


print(f"\n{'='*60}")
print("TRAINING BASELINE")
print(f"{'='*60}")
print(f"Epochs: {BASELINE_EPOCHS}")
print(f"Train batches: {len(rm_train_loader)}")
print(f"Val batches: {len(rm_val_loader)}")

baseline_history = {'epoch': [], 'train_loss': [], 'val_loss': [], 'val_dice': [], 'val_iou': []}
best_baseline_dice = 0.0

for epoch in range(1, BASELINE_EPOCHS + 1):
    start_time = time.time()

    train_loss, train_dice = train_baseline_epoch(
        baseline_model, rm_train_loader, baseline_criterion,
        baseline_optimizer, baseline_scaler, cfg, baseline_device
    )
    val_loss, val_dice, val_iou = validate_baseline(
        baseline_model, rm_val_loader, baseline_criterion, cfg, baseline_device
    )
    baseline_scheduler.step()

    print(f"\n{'='*60}")
    print(f"EPOCH {epoch}/{BASELINE_EPOCHS} | LR: {baseline_optimizer.param_groups[0]['lr']:.2e} | "
          f"Time: {time.time()-start_time:.1f}s")
    print(f"{'='*60}")
    print(f"Train — Loss: {train_loss:.4f} | Dice: {train_dice:.4f}")
    print(f"Val   — Loss: {val_loss:.4f} | Dice: {val_dice:.4f} | IoU: {val_iou:.4f}")

    baseline_history['epoch'].append(epoch)
    baseline_history['train_loss'].append(train_loss)
    baseline_history['val_loss'].append(val_loss)
    baseline_history['val_dice'].append(val_dice)
    baseline_history['val_iou'].append(val_iou)

    if val_dice > best_baseline_dice:
        best_baseline_dice = val_dice
        torch.save(
            baseline_model.state_dict(),
            os.path.join(cfg.CHECKPOINT_DIR, "baseline_unet_best.pt")
        )

history_path = os.path.join(cfg.LOG_DIR, 'baseline_history.json')
with open(history_path, 'w') as f:
    json.dump(baseline_history, f, indent=2)
print(f"\nSaved history: {history_path}")

# ============================================================
# 12.6 PLOT & DIAGNOSIS
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(baseline_history['epoch'], baseline_history['train_loss'], label='Train Loss', marker='o')
axes[0].plot(baseline_history['epoch'], baseline_history['val_loss'], label='Val Loss', marker='o')
axes[0].set_xlabel('Epoch')
axes[0].set_ylabel('Loss')
axes[0].set_title('Baseline U-Net Loss')
axes[0].legend()
axes[0].grid(True)

axes[1].plot(baseline_history['epoch'], baseline_history['val_dice'], label='Val Dice', marker='o', color='green')
axes[1].axhline(y=0.27, color='red', linestyle='--', label='TopoEyeNet ceiling (~0.27)')
axes[1].axhline(y=0.75, color='gray', linestyle=':', label='Typical DRIVE baseline (~0.75)')
axes[1].set_xlabel('Epoch')
axes[1].set_ylabel('Dice Score')
axes[1].set_title('Baseline U-Net Validation Dice')
axes[1].legend()
axes[1].grid(True)

plt.suptitle("Diagnostic Baseline: Vanilla U-Net on RetinoMix-5", fontsize=14, fontweight='bold')
plt.tight_layout()
save_path = os.path.join(cfg.FIGURES_DIR, "baseline_diagnostic_curves.png")
plt.savefig(save_path, dpi=150, bbox_inches='tight')
print(f"\nSaved: {save_path}")
plt.show()

print(f"\n{'='*60}")
print("DIAGNOSIS")
print(f"{'='*60}")
print(f"  Baseline best val Dice:      {best_baseline_dice:.4f}")
print(f"  TopoEyeNet (v2) best Dice:   ~0.22-0.24 (from Cell 9)")
print(f"  TopoEyeNet (pretrained) Dice: ~0.24-0.27 so far (from Cell 11)")
print()
if best_baseline_dice < 0.35:
    print("  -> Baseline ALSO plateaus low.")
    print("     This points to an upstream data/pipeline issue:")
    print("     check image resolution vs vessel width, mask binarization,")
    print("     augmentation correctness, and train/val split size (163 images is small).")
elif best_baseline_dice > 0.6:
    print("  -> Baseline clears a healthy Dice range while TopoEyeNet does not.")
    print("     This points to a TopoEyeNet-specific issue:")
    print("     DataParallel batch-size-2 BatchNorm instability, the multi-task")
    print("     loss balancing, or the structural consistency term interfering")
    print("     with mask learning.")
else:
    print("  -> Baseline is better than TopoEyeNet but still below typical")
    print("     literature numbers. Likely BOTH a moderate data/pipeline issue")
    print("     AND a TopoEyeNet-specific issue are present. Worth extending")
    print("     this baseline run longer (this was only 20 epochs) before")
    print("     concluding further.")

get_gpu_memory()
print(f"\n{'='*60}")
print("CELL 12 COMPLETE")
print(f"{'='*60}")
```

```python
# ============================================================
# COMPREHENSIVE EVALUATION: ALL METRICS ON HELD-OUT TEST SET
# ============================================================
# Computes Dice, IoU, Accuracy, Sensitivity, Specificity, Precision,
# and AUC-ROC for both TopoEyeNet (pretrained, best checkpoint) and
# the vanilla U-Net baseline, on a TRUE held-out test set (not the
# validation set used for checkpoint selection -- reporting on val
# would inflate numbers and is not defensible in a paper).

import torch
import numpy as np
from sklearn.metrics import roc_auc_score
import segmentation_models_pytorch as smp

print("=" * 60)
print("COMPREHENSIVE EVALUATION SETUP")
print("=" * 60)

# --- Get a genuine test loader, distinct from rm_val_loader ---
try:
    test_loader = rm_test_loader
    print("Using existing rm_test_loader.")
except NameError:
    print("WARNING: rm_test_loader not found in scope.")
    print("Falling back to rebuilding the test split from retinomix.pairs")
    print("via split_dataset(). If this also fails, evaluation below")
    print("will use rm_val_loader as a LAST RESORT -- flag this in the")
    print("paper as a limitation if so, since val was used for model selection.")
    try:
        train_idx, val_idx, test_idx = split_dataset(retinomix.pairs, seed=cfg.SEED)
        test_pairs = [retinomix.pairs[i] for i in test_idx]
        test_dataset = RetinoMixDataset(test_pairs, cfg, split='test')  # adjust class name if different
        test_loader = torch.utils.data.DataLoader(
            test_dataset, batch_size=cfg.BATCH_SIZE, shuffle=False,
            num_workers=cfg.NUM_WORKERS, pin_memory=True
        )
        print(f"Rebuilt test loader: {len(test_pairs)} samples")
    except Exception as e:
        print(f"Could not rebuild test split ({e}); using rm_val_loader as fallback.")
        test_loader = rm_val_loader

device_eval = torch.device("cuda:0")

# ============================================================
# METRIC COMPUTATION
# ============================================================

@torch.no_grad()
def compute_full_metrics(model, loader, device, is_multitask=True, threshold=0.5):
    """
    Runs the model over the full loader and computes pixel-level
    metrics across the ENTIRE test set (not averaged per-batch --
    accumulated confusion-matrix style, which is the correct way
    to compute Sensitivity/Specificity/AUC across an imbalanced class).
    """
    model.eval()

    tp = fp = tn = fn = 0
    all_probs = []
    all_targets = []

    for images, targets in loader:
        images = images.to(device, non_blocking=True)
        mask_target = targets['mask'].to(device, non_blocking=True)

        if is_multitask:
            outputs = model(images)
            logits = outputs['mask']
        else:
            logits = model(images)

        probs = torch.sigmoid(logits)
        preds = (probs > threshold).float()

        tp += ((preds == 1) & (mask_target == 1)).sum().item()
        tn += ((preds == 0) & (mask_target == 0)).sum().item()
        fp += ((preds == 1) & (mask_target == 0)).sum().item()
        fn += ((preds == 0) & (mask_target == 1)).sum().item()

        # Subsample for AUC to keep memory sane (every 4th pixel)
        all_probs.append(probs.flatten()[::4].cpu().numpy())
        all_targets.append(mask_target.flatten()[::4].cpu().numpy())

    eps = 1e-7
    accuracy = (tp + tn) / (tp + tn + fp + fn + eps)
    sensitivity = tp / (tp + fn + eps)          # recall / TPR
    specificity = tn / (tn + fp + eps)          # TNR
    precision = tp / (tp + fp + eps)            # PPV
    dice = 2 * tp / (2 * tp + fp + fn + eps)    # == F1
    iou = tp / (tp + fp + fn + eps)

    all_probs = np.concatenate(all_probs)
    all_targets = np.concatenate(all_targets)
    try:
        auc = roc_auc_score(all_targets, all_probs)
    except Exception as e:
        auc = float('nan')
        print(f"  AUC computation failed: {e}")

    return {
        'accuracy': accuracy, 'sensitivity_recall': sensitivity,
        'specificity': specificity, 'precision': precision,
        'dice_f1': dice, 'iou': iou, 'auc_roc': auc,
        'tp': tp, 'fp': fp, 'tn': tn, 'fn': fn,
    }


def print_metrics(name, m):
    print(f"\n{'='*60}")
    print(f"{name}")
    print(f"{'='*60}")
    print(f"  Accuracy:              {m['accuracy']*100:.2f}%")
    print(f"  AUC-ROC:               {m['auc_roc']*100:.2f}%")
    print(f"  Specificity:           {m['specificity']*100:.2f}%")
    print(f"  Sensitivity (Recall):  {m['sensitivity_recall']*100:.2f}%")
    print(f"  Precision:             {m['precision']*100:.2f}%")
    print(f"  Dice / F1:             {m['dice_f1']*100:.2f}%")
    print(f"  IoU:                   {m['iou']*100:.2f}%")


# ============================================================
# RUN: TopoEyeNet (best checkpoint)
# ============================================================

print(f"\n{'='*60}")
print("LOADING TopoEyeNet BEST CHECKPOINT")
print(f"{'='*60}")

topo_eval_model = TopoEyeNetPretrained(
    encoder_name="resnet50",
    in_channels=cfg.IN_CHANNELS,
    mask_head=cfg.USE_MASK_HEAD,
    skeleton_head=cfg.USE_SKELETON_HEAD,
    pretrained_path=None,  # weights loaded from checkpoint below, not the MAE encoder
).to(device_eval)

ckpt = torch.load(
    os.path.join(cfg.CHECKPOINT_DIR, "topoeyenet_pretrained_best.pt"),
    map_location=device_eval
)
topo_eval_model.load_state_dict(ckpt['model_state_dict'])
print(f"Loaded checkpoint from epoch {ckpt['epoch']}")

topo_metrics = compute_full_metrics(topo_eval_model, test_loader, device_eval, is_multitask=True)
print_metrics("TopoEyeNet (pretrained encoder, best checkpoint)", topo_metrics)

# ============================================================
# RUN: Baseline U-Net
# ============================================================

print(f"\n{'='*60}")
print("LOADING BASELINE U-NET")
print(f"{'='*60}")

baseline_eval_model = smp.Unet(
    encoder_name="resnet34", encoder_weights=None, in_channels=3, classes=1,
).to(device_eval)
baseline_eval_model.load_state_dict(
    torch.load(os.path.join(cfg.CHECKPOINT_DIR, "baseline_unet_best.pt"), map_location=device_eval)
)

baseline_metrics = compute_full_metrics(baseline_eval_model, test_loader, device_eval, is_multitask=False)
print_metrics("Baseline U-Net (mask only)", baseline_metrics)

# ============================================================
# SIDE-BY-SIDE COMPARISON
# ============================================================

print(f"\n{'='*60}")
print("SIDE-BY-SIDE COMPARISON")
print(f"{'='*60}")
print(f"{'Metric':<25} {'TopoEyeNet':>12} {'Baseline':>12}")
print(f"{'-'*25} {'-'*12} {'-'*12}")
for key, label in [
    ('accuracy', 'Accuracy'), ('auc_roc', 'AUC-ROC'),
    ('specificity', 'Specificity'), ('sensitivity_recall', 'Sensitivity'),
    ('precision', 'Precision'), ('dice_f1', 'Dice/F1'), ('iou', 'IoU'),
]:
    print(f"{label:<25} {topo_metrics[key]*100:>11.2f}% {baseline_metrics[key]*100:>11.2f}%")

print(f"\n{'='*60}")
print("REALITY CHECK vs PUBLISHED SOTA (DRIVE/STARE/CHASE benchmarks)")
print(f"{'='*60}")
print("  Published SOTA typically shows:")
print("    Accuracy:     95-97%")
print("    AUC-ROC:      97-98%")
print("    Specificity:  97-98%")
print("    Sensitivity:  75-82%  <- usually the hardest to push")
print("    Dice/F1:      80-83%  <- inherently capped by class imbalance")
print()
print("  If your numbers above are noticeably below these ranges,")
print("  the biggest known levers (in order of typical impact) are:")
print("    1. Patch-based training instead of whole-image resize")
print("       (standard in this literature; turns ~160 images into")
print("       tens of thousands of training patches)")
print("    2. Preprocessing: green-channel extraction + CLAHE contrast")
print("       enhancement (vessels are far more visible to the model)")
print("    3. FOV masking: exclude background outside the retinal")
print("       circle from loss AND metrics (currently likely included,")
print("       which can distort Accuracy/Specificity upward artificially")
print("       and Dice/Sensitivity in unpredictable ways)")
print("    4. Test-time overlap-tile inference / augmentation averaging")

print(f"\n{'='*60}")
print("EVALUATION COMPLETE")
print(f"{'='*60}")
```

```python
# ============================================================
# THRESHOLD & POST-PROCESSING OPTIMIZATION (no retraining needed)
# ============================================================
# High Sensitivity (86%) + low Precision (61%) = the model is
# over-predicting vessel pixels. Two free levers, tuned on VAL
# (never on test, to keep the final test numbers honest):
#   1. Decision threshold -- 0.5 is rarely optimal for imbalanced masks
#   2. Small connected-component removal -- cleans up speckle false
#      positives that a flat threshold can't distinguish from real
#      thin vessels

import torch
import numpy as np
from skimage.morphology import remove_small_objects
import itertools

device_eval = torch.device("cuda:0")

# ============================================================
# 1. COLLECT RAW PROBABILITY MAPS + TARGETS ON VAL SET (once)
# ============================================================

@torch.no_grad()
def collect_val_predictions(model, loader, device, is_multitask=True):
    model.eval()
    all_probs, all_targets = [], []
    for images, targets in loader:
        images = images.to(device, non_blocking=True)
        mask_target = targets['mask'].to(device, non_blocking=True)

        if is_multitask:
            outputs = model(images)
            logits = outputs['mask']
        else:
            logits = model(images)

        probs = torch.sigmoid(logits).cpu().numpy()
        mask_np = mask_target.cpu().numpy()
        all_probs.append(probs)
        all_targets.append(mask_np)

    return all_probs, all_targets  # lists of (B,1,H,W) arrays, per batch


print("=" * 60)
print("COLLECTING VAL PREDICTIONS (TopoEyeNet)")
print("=" * 60)
val_probs, val_targets = collect_val_predictions(topo_eval_model, rm_val_loader, device_eval, is_multitask=True)
print(f"Collected {len(val_probs)} batches")

# ============================================================
# 2. THRESHOLD + MIN-SIZE GRID SEARCH (maximize Dice on VAL)
# ============================================================

def dice_precision_recall(pred_bin, target_bin):
    tp = np.logical_and(pred_bin == 1, target_bin == 1).sum()
    fp = np.logical_and(pred_bin == 1, target_bin == 0).sum()
    fn = np.logical_and(pred_bin == 0, target_bin == 1).sum()
    eps = 1e-7
    dice = 2 * tp / (2 * tp + fp + fn + eps)
    precision = tp / (tp + fp + eps)
    recall = tp / (tp + fn + eps)
    iou = tp / (tp + fp + fn + eps)
    return dice, precision, recall, iou


def apply_postprocess(prob_batch, threshold, min_size):
    """prob_batch: (B,1,H,W) numpy array of probabilities."""
    pred_bin = (prob_batch > threshold).astype(np.uint8)
    if min_size > 0:
        cleaned = np.zeros_like(pred_bin)
        for b in range(pred_bin.shape[0]):
            m = pred_bin[b, 0].astype(bool)
            m_clean = remove_small_objects(m, min_size=min_size)
            cleaned[b, 0] = m_clean.astype(np.uint8)
        pred_bin = cleaned
    return pred_bin


thresholds = [0.35, 0.40, 0.45, 0.50, 0.55, 0.60, 0.65, 0.70]
min_sizes = [0, 15, 30, 50]

print(f"\n{'='*60}")
print("GRID SEARCH: threshold x min_size (optimizing Dice on VAL)")
print(f"{'='*60}")
print(f"{'Threshold':>10} {'MinSize':>8} {'Dice':>8} {'Precision':>10} {'Recall':>8} {'IoU':>8}")

best_dice = -1
best_config = None
results = []

for thresh, min_size in itertools.product(thresholds, min_sizes):
    tp = fp = fn = 0
    for prob_batch, target_batch in zip(val_probs, val_targets):
        pred_bin = apply_postprocess(prob_batch, thresh, min_size)
        target_bin = (target_batch > 0.5).astype(np.uint8)
        tp += np.logical_and(pred_bin == 1, target_bin == 1).sum()
        fp += np.logical_and(pred_bin == 1, target_bin == 0).sum()
        fn += np.logical_and(pred_bin == 0, target_bin == 1).sum()

    eps = 1e-7
    dice = 2 * tp / (2 * tp + fp + fn + eps)
    precision = tp / (tp + fp + eps)
    recall = tp / (tp + fn + eps)
    iou = tp / (tp + fp + fn + eps)

    results.append((thresh, min_size, dice, precision, recall, iou))
    print(f"{thresh:>10.2f} {min_size:>8} {dice*100:>7.2f}% {precision*100:>9.2f}% {recall*100:>7.2f}% {iou*100:>7.2f}%")

    if dice > best_dice:
        best_dice = dice
        best_config = (thresh, min_size)

print(f"\n{'='*60}")
print(f"BEST CONFIG (on VAL): threshold={best_config[0]}, min_size={best_config[1]}")
print(f"  Val Dice at best config: {best_dice*100:.2f}%")
print(f"{'='*60}")

# ============================================================
# 3. APPLY BEST CONFIG TO TEST SET (final, honest numbers)
# ============================================================

print(f"\n{'='*60}")
print("APPLYING BEST CONFIG TO TEST SET")
print(f"{'='*60}")

best_thresh, best_min_size = best_config

test_probs, test_targets = collect_val_predictions(topo_eval_model, rm_test_loader, device_eval, is_multitask=True)

tp = fp = fn = tn = 0
for prob_batch, target_batch in zip(test_probs, test_targets):
    pred_bin = apply_postprocess(prob_batch, best_thresh, best_min_size)
    target_bin = (target_batch > 0.5).astype(np.uint8)
    tp += np.logical_and(pred_bin == 1, target_bin == 1).sum()
    fp += np.logical_and(pred_bin == 1, target_bin == 0).sum()
    fn += np.logical_and(pred_bin == 0, target_bin == 1).sum()
    tn += np.logical_and(pred_bin == 0, target_bin == 0).sum()

eps = 1e-7
final_accuracy = (tp + tn) / (tp + tn + fp + fn + eps)
final_sensitivity = tp / (tp + fn + eps)
final_specificity = tn / (tn + fp + eps)
final_precision = tp / (tp + fp + eps)
final_dice = 2 * tp / (2 * tp + fp + fn + eps)
final_iou = tp / (tp + fp + fn + eps)

print(f"  Accuracy:              {final_accuracy*100:.2f}%")
print(f"  Specificity:           {final_specificity*100:.2f}%")
print(f"  Sensitivity (Recall):  {final_sensitivity*100:.2f}%")
print(f"  Precision:             {final_precision*100:.2f}%")
print(f"  Dice / F1:             {final_dice*100:.2f}%")
print(f"  IoU:                   {final_iou*100:.2f}%")
print(f"  (AUC-ROC unaffected by threshold -- unchanged from prior eval)")

print(f"\n{'='*60}")
print("BEFORE vs AFTER (test set)")
print(f"{'='*60}")
print(f"{'Metric':<15} {'Before (thr=0.5)':>18} {'After (tuned)':>16}")
print(f"{'Precision':<15} {'61.43%':>18} {final_precision*100:>15.2f}%")
print(f"{'Dice/F1':<15} {'71.81%':>18} {final_dice*100:>15.2f}%")
print(f"{'IoU':<15} {'56.02%':>18} {final_iou*100:>15.2f}%")

print(f"\n{'='*60}")
print("REALITY CHECK")
print(f"{'='*60}")
if final_dice >= 0.80:
    print("  Dice crossed 80% from threshold/postprocessing tuning alone.")
else:
    print(f"  Dice improved but likely still below 80% (this lever typically")
    print(f"  buys a few points, not 8-10). To close the remaining gap, the")
    print(f"  next highest-impact changes (in order) are:")
    print(f"    1. Patch-based training (biggest single lever in this")
    print(f"       literature -- turns ~160 images into tens of thousands")
    print(f"       of training patches, dramatically more signal per epoch)")
    print(f"    2. CLAHE contrast enhancement + green-channel preprocessing")
    print(f"       (standard in retinal vessel segmentation, makes thin")
    print(f"       vessels far more separable from background)")
    print(f"    3. FOV masking -- exclude pixels outside the retinal disc")
    print(f"       from both loss and metrics; background-corner false")
    print(f"       positives may be inflating your FP count right now")
```

```python
# ============================================================
# clDice: TOPOLOGY-PRESERVING LOSS + METRIC
# ============================================================
# Reference: Shit et al., "clDice -- A Novel Topology-Preserving Loss
# Function for Tubular Structure Segmentation", CVPR 2021.
#
# Standard clDice derives a soft skeleton from the predicted MASK via
# differentiable morphological thinning. TopoEyeNet already has an
# explicit skeleton HEAD, so this implementation uses a simplified,
# architecture-native variant: instead of deriving the skeleton via
# soft-skeletonization, it uses your model's actual predicted skeleton
# output directly. This is cheaper (no iterative erosion/dilation) and
# also doubles as a direct check of mask/skeleton cross-consistency --
# which is exactly your "structural consistency" claim.
#
# tprec = fraction of PREDICTED skeleton that lies inside the TRUE mask
# tsens = fraction of the TRUE skeleton that is covered by the PREDICTED mask
# clDice = harmonic mean of tprec and tsens (topology-aware "Dice")
#
# A full soft-skeletonization version (the literature-standard, mask-only
# variant with no separate skeleton head required) is also included
# below for use as an additional evaluation metric / ablation, since
# reviewers familiar with clDice will expect to see the standard form
# reported alongside your architecture-native one.

import torch
import torch.nn as nn
import torch.nn.functional as F


# ============================================================
# 1. ARCHITECTURE-NATIVE clDice LOSS (uses your existing dual heads)
# ============================================================

class ClDiceLoss(nn.Module):
    """
    Topology-aware loss using TopoEyeNet's existing mask + skeleton
    heads directly -- no soft-skeletonization needed.
    """
    def __init__(self, smooth: float = 1.0):
        super().__init__()
        self.smooth = smooth

    def forward(self, mask_logits, skeleton_logits, mask_target, skeleton_target):
        mask_prob = torch.sigmoid(mask_logits)
        skel_prob = torch.sigmoid(skeleton_logits)

        # Precision: how much of the PREDICTED skeleton lands inside the TRUE mask
        tprec = (torch.sum(skel_prob * mask_target) + self.smooth) / \
                (torch.sum(skel_prob) + self.smooth)

        # Sensitivity: how much of the TRUE skeleton is covered by the PREDICTED mask
        tsens = (torch.sum(skeleton_target * mask_prob) + self.smooth) / \
                (torch.sum(skeleton_target) + self.smooth)

        cl_dice = 1.0 - 2.0 * (tprec * tsens) / (tprec + tsens + self.smooth)
        return cl_dice


# ============================================================
# 2. STANDARD (soft-skeletonization) clDice -- for reporting/ablation
# ============================================================

def soft_erode(img):
    if img.dim() == 4:
        p1 = -F.max_pool2d(-img, (3, 1), (1, 1), (1, 0))
        p2 = -F.max_pool2d(-img, (1, 3), (1, 1), (0, 1))
        return torch.min(p1, p2)
    raise ValueError("Expected 4D tensor (B,C,H,W)")

def soft_dilate(img):
    return F.max_pool2d(img, (3, 3), (1, 1), (1, 1))

def soft_open(img):
    return soft_dilate(soft_erode(img))

def soft_skeletonize(img, iterations: int = 10):
    """Differentiable morphological thinning -> soft skeleton in [0,1]."""
    img1 = soft_open(img)
    skel = F.relu(img - img1)
    for _ in range(iterations):
        img = soft_erode(img)
        img1 = soft_open(img)
        delta = F.relu(img - img1)
        skel = skel + F.relu(delta - skel * delta)
    return skel

def standard_cldice_metric(pred_mask_prob, true_mask_binary, iterations: int = 10, smooth: float = 1e-6):
    """
    Standard literature-form clDice, computed mask-only (no skeleton
    head needed). Use this to report a number directly comparable to
    other clDice results in the vessel-segmentation literature.
    """
    skel_pred = soft_skeletonize(pred_mask_prob, iterations)
    skel_true = soft_skeletonize(true_mask_binary, iterations)

    tprec = (torch.sum(skel_pred * true_mask_binary) + smooth) / (torch.sum(skel_pred) + smooth)
    tsens = (torch.sum(skel_true * pred_mask_prob) + smooth) / (torch.sum(skel_true) + smooth)

    cl_dice = 2.0 * (tprec * tsens) / (tprec + tsens + smooth)
    return cl_dice.item(), tprec.item(), tsens.item()


# ============================================================
# 3. INTEGRATION: drop into TopoEyeLossV2
# ============================================================
# Add this to TopoEyeLossV2.__init__:
#     self.cldice = ClDiceLoss(smooth=1.0)
#     self.lambda_cldice = 0.5   # tune this
#
# Add this inside TopoEyeLossV2.forward, after both mask and skeleton
# losses are computed:
#     if 'mask' in predictions and 'skeleton' in predictions:
#         losses['cldice'] = self.cldice(
#             predictions['mask'], predictions['skeleton'],
#             targets['mask'], targets['skeleton']
#         )
#         total_loss += self.lambda_cldice * losses['cldice']
#
# This is IN ADDITION to your existing MaskSkeletonAgreementLoss
# (consistency term) -- they measure different things. Agreement loss
# penalizes pixel-wise mask/skeleton mismatch; clDice specifically
# rewards correct CONNECTIVITY/topology even under small positional
# offset, which plain pixel losses cannot do.

print("clDice loss + metric defined.")
print("See the INTEGRATION comment block above for how to wire ClDiceLoss")
print("into TopoEyeLossV2 -- requires re-running Cell 6/9's loss definition")
print("and then re-running the fine-tuning cell (Cell 11) to actually train")
print("with it, since this is a training-time change, not just an eval one.")
```

```python
# ============================================================
# PATCH-BASED TRAINING PIPELINE (CLAHE + FOV masking)
# ============================================================
# Standard high-Dice recipe in retinal vessel segmentation literature:
#   1. Preprocess: green channel + CLAHE contrast enhancement
#   2. FOV mask: exclude background outside the retinal disc
#   3. Extract many overlapping patches per image (native resolution,
#      no downsampling loss), oversampling vessel-rich regions
#   4. Train on patches (effectively tens of thousands of samples
#      instead of ~160 whole images)
#   5. At inference: overlap-tile reconstruction back to full image,
#      so final Dice is still reported on whole images (comparable
#      to your existing numbers)

import numpy as np
import cv2
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import segmentation_models_pytorch as smp
from torch.cuda.amp import autocast, GradScaler
import random
import time

# ============================================================
# 1. PREPROCESSING: GREEN CHANNEL + CLAHE
# ============================================================

def preprocess_fundus(image_rgb: np.ndarray) -> np.ndarray:
    """
    Standard retinal vessel preprocessing:
    - Extract green channel (vessels have highest contrast here)
    - CLAHE contrast enhancement
    - Returns single-channel float32 [0,1], then replicated to 3ch
      so it's still compatible with a pretrained 3-channel encoder.
    """
    green = image_rgb[:, :, 1]
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    enhanced = clahe.apply(green)
    enhanced = enhanced.astype(np.float32) / 255.0
    # Replicate to 3 channels so pretrained encoders still work
    return np.stack([enhanced, enhanced, enhanced], axis=-1)


def compute_fov_mask(image_rgb: np.ndarray, threshold: int = 15) -> np.ndarray:
    """
    Simple FOV (field of view) mask: the retinal disc is bright,
    the surrounding background is near-black. Thresholding the
    green channel and taking the largest connected component gives
    a reliable FOV mask without needing per-dataset FOV files.
    """
    green = image_rgb[:, :, 1]
    _, binary = cv2.threshold(green, threshold, 255, cv2.THRESH_BINARY)
    binary = binary.astype(np.uint8)

    # Clean up: keep only the largest connected component, fill holes
    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(binary, connectivity=8)
    if num_labels > 1:
        largest = 1 + np.argmax(stats[1:, cv2.CC_STAT_AREA])
        fov = (labels == largest).astype(np.uint8)
    else:
        fov = binary

    # Erode slightly to avoid bright rim artifacts at the FOV edge
    kernel = np.ones((9, 9), np.uint8)
    fov = cv2.erode(fov, kernel, iterations=1)
    return fov


# ============================================================
# 2. PATCH DATASET (on-the-fly random extraction, oversamples vessels)
# ============================================================

class PatchDataset(Dataset):
    """
    Loads full-resolution images/masks lazily, extracts random patches
    on the fly each call (so effectively infinite augmentation across
    epochs). ~50% of patches are centered on vessel pixels (oversampling
    the minority class), the rest are random-but-within-FOV.
    """

    def __init__(self, image_paths, mask_paths, precomputed_dir,
                 patch_size=128, patches_per_image=40, vessel_oversample_ratio=0.5,
                 augment=True):
        self.image_paths = image_paths
        self.mask_paths = mask_paths
        self.precomputed_dir = precomputed_dir
        self.patch_size = patch_size
        self.patches_per_image = patches_per_image
        self.vessel_oversample_ratio = vessel_oversample_ratio
        self.augment = augment

        # Filter to images with valid precomputed masks (reuse existing convention)
        self.valid_indices = []
        for i, mp in enumerate(mask_paths):
            if mp is None:
                continue
            base = os.path.splitext(os.path.basename(mp))[0]
            npz_path = os.path.join(precomputed_dir, f"{base}_reps.npz")
            if os.path.exists(npz_path):
                self.valid_indices.append(i)

        print(f"PatchDataset: {len(self.valid_indices)} source images, "
              f"{patches_per_image} patches/image = "
              f"{len(self.valid_indices) * patches_per_image} patches/epoch")

    def __len__(self):
        return len(self.valid_indices) * self.patches_per_image

    def _load_image_and_mask(self, real_idx):
        img_path = self.image_paths[real_idx]
        mask_path = self.mask_paths[real_idx]

        image = np.array(Image.open(img_path).convert('RGB'))
        base = os.path.splitext(os.path.basename(mask_path))[0]
        npz_path = os.path.join(self.precomputed_dir, f"{base}_reps.npz")
        reps = dict(np.load(npz_path))
        mask = reps['mask'].astype(np.float32)

        h, w = image.shape[:2]
        if mask.shape[0] != h or mask.shape[1] != w:
            mask = cv2.resize(mask, (w, h), interpolation=cv2.INTER_NEAREST)

        preprocessed = preprocess_fundus(image)
        fov = compute_fov_mask(image)

        return preprocessed, mask, fov

    def __getitem__(self, idx):
        real_idx = self.valid_indices[idx % len(self.valid_indices)]
        image, mask, fov = self._load_image_and_mask(real_idx)
        h, w = mask.shape
        ps = self.patch_size

        vessel_ys, vessel_xs = np.where((mask > 0) & (fov > 0))
        use_vessel_center = (
            len(vessel_ys) > 0 and random.random() < self.vessel_oversample_ratio
        )

        for _ in range(20):  # a few retries to land a valid in-FOV patch
            if use_vessel_center:
                i = random.randint(0, len(vessel_ys) - 1)
                cy, cx = vessel_ys[i], vessel_xs[i]
            else:
                cy = random.randint(ps // 2, h - ps // 2 - 1)
                cx = random.randint(ps // 2, w - ps // 2 - 1)

            y0 = max(0, min(h - ps, cy - ps // 2))
            x0 = max(0, min(w - ps, cx - ps // 2))

            fov_patch = fov[y0:y0 + ps, x0:x0 + ps]
            if fov_patch.mean() > 0.5:  # patch is mostly inside FOV
                break

        img_patch = image[y0:y0 + ps, x0:x0 + ps].copy()
        mask_patch = mask[y0:y0 + ps, x0:x0 + ps].copy()

        if self.augment:
            if random.random() < 0.5:
                img_patch = np.fliplr(img_patch).copy()
                mask_patch = np.fliplr(mask_patch).copy()
            if random.random() < 0.5:
                img_patch = np.flipud(img_patch).copy()
                mask_patch = np.flipud(mask_patch).copy()
            k = random.randint(0, 3)
            if k > 0:
                img_patch = np.rot90(img_patch, k).copy()
                mask_patch = np.rot90(mask_patch, k).copy()

        img_tensor = torch.from_numpy(img_patch.transpose(2, 0, 1)).float()
        mask_tensor = torch.from_numpy(mask_patch).unsqueeze(0).float()

        return img_tensor, mask_tensor


# ============================================================
# 3. BUILD PATCH LOADERS (reuse existing train/val split)
# ============================================================

PATCH_SIZE = 128
PATCHES_PER_IMAGE_TRAIN = 60
PATCHES_PER_IMAGE_VAL = 20

print("=" * 60)
print("BUILDING PATCH DATASETS")
print("=" * 60)

patch_train_dataset = PatchDataset(
    rm_train_images, rm_train_masks,
    precomputed_dir=os.path.join(cfg.PRECOMPUTED_DIR, "retinomix5"),
    patch_size=PATCH_SIZE, patches_per_image=PATCHES_PER_IMAGE_TRAIN,
    vessel_oversample_ratio=0.5, augment=True,
)
patch_val_dataset = PatchDataset(
    rm_val_images, rm_val_masks,
    precomputed_dir=os.path.join(cfg.PRECOMPUTED_DIR, "retinomix5"),
    patch_size=PATCH_SIZE, patches_per_image=PATCHES_PER_IMAGE_VAL,
    vessel_oversample_ratio=0.5, augment=False,
)

PATCH_BATCH_SIZE = 32  # patches are small, can afford a much bigger batch than whole images

patch_train_loader = DataLoader(
    patch_train_dataset, batch_size=PATCH_BATCH_SIZE, shuffle=True,
    num_workers=cfg.NUM_WORKERS, pin_memory=True, drop_last=True
)
patch_val_loader = DataLoader(
    patch_val_dataset, batch_size=PATCH_BATCH_SIZE, shuffle=False,
    num_workers=cfg.NUM_WORKERS, pin_memory=True
)

print(f"Train: {len(patch_train_dataset)} patches/epoch, {len(patch_train_loader)} batches")
print(f"Val:   {len(patch_val_dataset)} patches/epoch, {len(patch_val_loader)} batches")

# ============================================================
# 4. TRAIN A U-NET ON PATCHES
# ============================================================

patch_device = torch.device("cuda:0")

patch_model = smp.Unet(
    encoder_name="resnet34", encoder_weights="imagenet", in_channels=3, classes=1,
).to(patch_device)

patch_criterion = DiceBCELoss(dice_weight=0.7, bce_weight=0.3)
patch_optimizer = torch.optim.AdamW(patch_model.parameters(), lr=1e-4, weight_decay=1e-5)
PATCH_EPOCHS = 30
patch_scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(patch_optimizer, T_max=PATCH_EPOCHS, eta_min=1e-6)
patch_scaler = GradScaler()

def dice_from_logits(logits, target, threshold=0.5):
    probs = torch.sigmoid(logits)
    preds = (probs > threshold).float()
    tp = (preds * target).sum()
    fp = (preds * (1 - target)).sum()
    fn = ((1 - preds) * target).sum()
    return (2 * tp / (2 * tp + fp + fn + 1e-7)).item()

print(f"\n{'='*60}")
print(f"TRAINING PATCH-BASED U-NET — {PATCH_EPOCHS} epochs")
print(f"{'='*60}")

best_patch_dice = 0.0
for epoch in range(1, PATCH_EPOCHS + 1):
    patch_model.train()
    t0 = time.time()
    train_loss, train_dice, n = 0.0, 0.0, 0

    for images, masks in patch_train_loader:
        images = images.to(patch_device, non_blocking=True)
        masks = masks.to(patch_device, non_blocking=True)

        with autocast(enabled=True):
            logits = patch_model(images)
            loss = patch_criterion(logits, masks)

        patch_scaler.scale(loss).backward()
        patch_scaler.step(patch_optimizer)
        patch_scaler.update()
        patch_optimizer.zero_grad()

        train_loss += loss.item()
        train_dice += dice_from_logits(logits, masks)
        n += 1

    patch_scheduler.step()
    train_loss /= n
    train_dice /= n

    patch_model.eval()
    val_dice, n_val = 0.0, 0
    with torch.no_grad():
        for images, masks in patch_val_loader:
            images = images.to(patch_device, non_blocking=True)
            masks = masks.to(patch_device, non_blocking=True)
            with autocast(enabled=True):
                logits = patch_model(images)
            val_dice += dice_from_logits(logits, masks)
            n_val += 1
    val_dice /= n_val

    if val_dice > best_patch_dice:
        best_patch_dice = val_dice
        torch.save(patch_model.state_dict(), os.path.join(cfg.CHECKPOINT_DIR, "patch_unet_best.pt"))

    print(f"Epoch {epoch}/{PATCH_EPOCHS} | Time {time.time()-t0:.1f}s | "
          f"Train Loss {train_loss:.4f} Dice {train_dice:.4f} | "
          f"Val Dice {val_dice:.4f} | LR {patch_optimizer.param_groups[0]['lr']:.2e}")

print(f"\nBest patch-level val Dice: {best_patch_dice:.4f}")
print(f"Saved: {cfg.CHECKPOINT_DIR}/patch_unet_best.pt")
print(f"\n{'='*60}")
print("NOTE: this is PATCH-level Dice, not full-image Dice.")
print("Run the overlap-tile inference script next to get full-image")
print("Dice comparable to your existing test-set numbers.")
print(f"{'='*60}")
```

```python
# ============================================================
# OVERLAP-TILE INFERENCE: full-image Dice from the patch-trained model
# ============================================================
# Slides the trained patch model across each full test image with
# overlapping windows, averages overlapping predictions, and computes
# Dice on the full reconstructed mask -- directly comparable to your
# existing whole-image test metrics.

import torch
import numpy as np
import cv2
from sklearn.metrics import roc_auc_score

patch_device = torch.device("cuda:0")

# Reload best patch model (in case running in a fresh session)
patch_model_eval = smp.Unet(
    encoder_name="resnet34", encoder_weights=None, in_channels=3, classes=1,
).to(patch_device)
patch_model_eval.load_state_dict(
    torch.load(os.path.join(cfg.CHECKPOINT_DIR, "patch_unet_best.pt"), map_location=patch_device)
)
patch_model_eval.eval()

PATCH_SIZE = 128
STRIDE = 64  # 50% overlap


@torch.no_grad()
def predict_full_image(model, image_rgb, patch_size=PATCH_SIZE, stride=STRIDE, batch_size=32):
    """Overlap-tile prediction: returns a full-resolution probability map."""
    preprocessed = preprocess_fundus(image_rgb)  # (H, W, 3) float32
    h, w = preprocessed.shape[:2]

    # Pad so patches divide evenly
    pad_h = (patch_size - h % stride) % stride if h % stride != 0 else 0
    pad_w = (patch_size - w % stride) % stride if w % stride != 0 else 0
    padded = np.pad(preprocessed, ((0, pad_h + patch_size), (0, pad_w + patch_size), (0, 0)), mode='reflect')
    ph, pw = padded.shape[:2]

    prob_sum = np.zeros((ph, pw), dtype=np.float32)
    count = np.zeros((ph, pw), dtype=np.float32)

    coords = []
    for y in range(0, ph - patch_size + 1, stride):
        for x in range(0, pw - patch_size + 1, stride):
            coords.append((y, x))

    for i in range(0, len(coords), batch_size):
        batch_coords = coords[i:i + batch_size]
        batch_patches = np.stack([
            padded[y:y + patch_size, x:x + patch_size] for y, x in batch_coords
        ])
        batch_tensor = torch.from_numpy(batch_patches.transpose(0, 3, 1, 2)).float().to(patch_device)

        with autocast(enabled=True):
            logits = model(batch_tensor)
        probs = torch.sigmoid(logits).cpu().numpy()[:, 0]

        for (y, x), p in zip(batch_coords, probs):
            prob_sum[y:y + patch_size, x:x + patch_size] += p
            count[y:y + patch_size, x:x + patch_size] += 1

    prob_map = prob_sum / np.maximum(count, 1e-7)
    return prob_map[:h, :w]  # crop back to original size


# ============================================================
# RUN ON FULL TEST SET
# ============================================================

print("=" * 60)
print("OVERLAP-TILE INFERENCE ON TEST SET (full images)")
print("=" * 60)

tp = fp = fn = tn = 0
all_probs_sample, all_targets_sample = [], []

test_pairs_list = list(zip(rm_test_images, rm_test_masks))
print(f"Evaluating {len(test_pairs_list)} full test images...")

for idx, (img_path, mask_path) in enumerate(test_pairs_list):
    image = np.array(Image.open(img_path).convert('RGB'))
    base = os.path.splitext(os.path.basename(mask_path))[0]
    npz_path = os.path.join(cfg.PRECOMPUTED_DIR, "retinomix5", f"{base}_reps.npz")
    if not os.path.exists(npz_path):
        continue
    reps = dict(np.load(npz_path))
    mask = reps['mask'].astype(np.float32)
    h, w = image.shape[:2]
    if mask.shape[0] != h or mask.shape[1] != w:
        mask = cv2.resize(mask, (w, h), interpolation=cv2.INTER_NEAREST)

    fov = compute_fov_mask(image)
    prob_map = predict_full_image(patch_model_eval, image)

    pred_bin = (prob_map > 0.5).astype(np.uint8)
    target_bin = (mask > 0.5).astype(np.uint8)

    # Restrict to FOV -- outside-FOV pixels shouldn't count either way
    valid = fov > 0

    tp += np.logical_and(pred_bin[valid] == 1, target_bin[valid] == 1).sum()
    fp += np.logical_and(pred_bin[valid] == 1, target_bin[valid] == 0).sum()
    fn += np.logical_and(pred_bin[valid] == 0, target_bin[valid] == 1).sum()
    tn += np.logical_and(pred_bin[valid] == 0, target_bin[valid] == 0).sum()

    all_probs_sample.append(prob_map[valid][::10])
    all_targets_sample.append(target_bin[valid][::10])

    if (idx + 1) % 5 == 0:
        print(f"  {idx+1}/{len(test_pairs_list)} images processed")

eps = 1e-7
final_accuracy = (tp + tn) / (tp + tn + fp + fn + eps)
final_sensitivity = tp / (tp + fn + eps)
final_specificity = tn / (tn + fp + eps)
final_precision = tp / (tp + fp + eps)
final_dice = 2 * tp / (2 * tp + fp + fn + eps)
final_iou = tp / (tp + fp + fn + eps)

all_probs_flat = np.concatenate(all_probs_sample)
all_targets_flat = np.concatenate(all_targets_sample)
final_auc = roc_auc_score(all_targets_flat, all_probs_flat)

print(f"\n{'='*60}")
print("PATCH-TRAINED MODEL — FULL-IMAGE TEST METRICS (FOV-masked)")
print(f"{'='*60}")
print(f"  Accuracy:              {final_accuracy*100:.2f}%")
print(f"  AUC-ROC:               {final_auc*100:.2f}%")
print(f"  Specificity:           {final_specificity*100:.2f}%")
print(f"  Sensitivity (Recall):  {final_sensitivity*100:.2f}%")
print(f"  Precision:             {final_precision*100:.2f}%")
print(f"  Dice / F1:             {final_dice*100:.2f}%")
print(f"  IoU:                   {final_iou*100:.2f}%")

print(f"\n{'='*60}")
print("COMPARISON: whole-image (previous) vs patch-based (this run)")
print(f"{'='*60}")
print(f"{'Metric':<15} {'Whole-image':>14} {'Patch-based':>14}")
print(f"{'Precision':<15} {'61.43%':>14} {final_precision*100:>13.2f}%")
print(f"{'Dice/F1':<15} {'71.81%':>14} {final_dice*100:>13.2f}%")
print(f"{'IoU':<15} {'56.02%':>14} {final_iou*100:>13.2f}%")

print(f"\n{'='*60}")
print("REALITY CHECK")
print(f"{'='*60}")
if final_dice >= 0.85:
    print("  85%+ Dice achieved.")
elif final_dice >= 0.80:
    print("  80%+ Dice achieved -- solidly in published SOTA range.")
    print("  Remaining gap to 85% would likely need: ensembling multiple")
    print("  seeds/architectures, deeper patch-model capacity, or more")
    print("  aggressive test-time augmentation (flips/rotations averaged).")
else:
    print("  Still below 80%. Next things to check, in order:")
    print("    1. Increase PATCHES_PER_IMAGE_TRAIN and PATCH_EPOCHS --")
    print("       this run used modest values to keep iteration fast.")
    print("    2. Verify FOV mask quality visually on a few images --")
    print("       a bad FOV mask silently corrupts both training sampling")
    print("       and evaluation.")
    print("    3. Try a larger patch size (e.g. 192 or 256) if vessels")
    print("       need more spatial context than 128px provides.")
```
